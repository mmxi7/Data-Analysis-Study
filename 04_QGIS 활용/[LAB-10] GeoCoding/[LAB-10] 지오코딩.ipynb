### #02 지오코딩

1. 브이월드 API를 사용하여 주소를 위경도로 변환

```python
geo_df = my_gis.geocode(df,                                             # 원본 데이터 프레임
                        addr="주소",                                     # 주소 컬럼명
                        key="24273A54-34F0-4A22-B01A-8052A3EB8E90")     # 브이월드에서 발급받은 API Key

geo_df.head()
```

2. 변환된 데이터 프레임 저장하기

```python
geo_df.to_excel("서울시_응급실_위치정보_geocoder.xlsx", index=False)
```
