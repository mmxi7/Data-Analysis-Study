## 2. CSV 파일 다루기

### #01 딕셔너리를 포함하는 리스트 구조를 CSV로 저장하기

```python
# 한 학급의 성적표 데이터 예시
grade = [
    {"name": "철수", "kor": 95, "eng": 88, "math": 72},
    {"name": "영희", "kor": 92, "eng": 90, "math": 95},
    {"name": "철민", "kor": 88, "eng": 76, "math": 64}
]

# CSV 파일의 한 행에 대한 구문 형식 정리
tpl = "{0},{1},{2},{3}\n"

# csv 파일 저장을 위한 f 객체 생성
# -> Excel의 csv는 euc-kr 형식
with open("grade.csv", "w", encoding = 'euc-kr') as f:
    # 첫 줄에 각 항목의 제목 기록
    f.write("이름,국어,영어,수학\n")

    # 각 데이터를 한 줄씩 콤마로 구분하여 기록
    for item in grade:
        tmp = tpl.format(item["name"], item["kor"], item["eng"], item["math"])
        f.write(tmp)
```

### #02 리스트를 포함하는 딕셔너리를 CSV로 저장하기

```python
grade_dic = {
    '이름': ['철수', '영희', '민철', '수현', '호영'],
    '국어': [98, 88, 92, 63, 120],
    '영어': [76, 90, 70, 60, 50],
    '수학': [88, 62, 88, 31, 76],
    '과학': [64, 72, 45, 70, 88]
}

tpl = "{0},{1},{2},{3}\n"
keys = list(grade_dic.keys())
p = ","
title = p.join(keys)

with open("grade.csv", "w", encoding = 'euc-kr') as f:
    f.write(title + "\n")

    for i in range(0, len(grade_dic['이름'])):
        tmp = tpl.format(grade_dic['이름'][i], grade_dic['국어'][i], grade_dic['영어'][i], grade_dic['수학'][i], grade_dic['과학'][i])
        f.write(tmp)
```

### #03 CSV 파일의 내용을 파이썬 데이터로 변환

```python
#------------------------------------------
# step1. 파일 읽기
#------------------------------------------
# 읽어들인 내용을 저장하기 위한 빈 변수 생성
csv_list = None

# grade.csv 파일에 대한 파일 객체를 읽기 모드로 생성
with open("grade.csv", "r", encoding = 'euc-kr') as f:
    csv_list = f.readlines()

print(csv_list)

#------------------------------------------
# step2. 읽어온 내용을 변환하기
#------------------------------------------
data = []

for i, line in enumerate(csv_list):
    if i == 0:              # 첫 행은 제목이므로 건너뛴다
        continue

    k = line.strip().split(",")
    item = {"name": k[0], "kor": int(k[1]), "eng": int(k[2]), "math": int(k[3])}
    data.append(item)

data
```

### #04 covid19로 인한 서울시와 전국 단위 일평균 확진/사망자 수

```python
try:
    with open("covid19.csv", 'r', encoding = 'euc-kr') as f:
        csv_list = f.readlines()
        # for i in range(5):
        #    print(csv_list[i].strip())

        # 전체 데이터 길이 (제목명을 제외하기 위해 - 1 처리)
        size = len(csv_list) - 1

        # 합계를 위한 변수 초기화
        seoul_confirm_sum = 0           # 서울 확진자 수 합계
        seoul_death_sum = 0             # 서울 사망자 수 합계
        nation_confirm_sum = 0          # 전국 확진자 수 합계
        nation_death_sum = 0            # 전국 사망자 수 합계

        for i, line in enumerate(csv_list):
            if i == 0:              # 첫 행은 제목이므로 건너뛴다
                continue

            k = line.strip().split(",")

            # 서울의 확진자 수와 사망자 수 추출
            if k[1].isnumeric():        seoul_confirm_sum += int(k[1])
            if k[2].isnumeric():        seoul_death_sum += int(k[2])

            # 전국의 확진자 수와 사망자 수 추출
            if k[3].isnumeric():        nation_confirm_sum += int(k[3])
            if k[4].isnumeric():        nation_death_sum += int(k[4])

        # 일 평균 출력
        print("서울 일 평균 확진자 수 : %.2f명" % (seoul_confirm_sum / size))
        print("서울 일 평균 사망자 수 : %.2f명" % (seoul_death_sum / size))
        print("서울 일 평균 확진자 수 : %.2f명" % (nation_confirm_sum / size))
        print("서울 일 평균 사망자 수 : %.2f명" % (nation_death_sum / size))

except Exception as e:
    print("CSV 파일 처리에 실패했습니다.", e)
finally:
    print("-" * 30)
    print("작업이 완료되었습니다.")
```
