## 4. 날짜 및 시간 관련 기능

### #01 현재 시각을 저장하고 있는 객체 생성

1. 라이브러리 참조

```python
import datetime as dt
```

2. 현재 시각을 갖는 객체 생성하기

```python
now_time = dt.datetime.now()
print(now_time)
```

### #02 날짜 객체의 성분 추출

1. 년, 월, 일 성분 추출

```python
yy = now_time.year
mm = now_time.month
dd = now_time.day

msg = "오늘은 %d년 %d월 %d일 입니다."
print(msg % (yy, mm, dd))
```

2. 시, 분, 초 성분 추출

```python
hh = now_time.hour
mi = now_time.minute
ss = now_time.second

msg = "지금은 %d시 %d분 %d초 입니다."
print(msg % (hh, mi, ss))
```

3. 현재 요일의 인덱스 조회

```python
d = now_time.weekday()
print(d)

# 조회한 요일 인덱스를 튜플에 적용하여 요일 이름 가져오기
days = ("월", "화", "수", "목", "금", "토", "일")
print(days[d])
```

4. 날짜 형식에 대한 형식 문자열 활용

```python
print(now_time.strftime("%y-%m-%d %H:%M:%S"))
```

5. 날짜 형식의 문자열을 날짜 객체로 변환

```python
print(now_time.strftime("%Y년 %m월 %d일 %H시 %M분 %S초"))
```

### #03 특정 시각 지정하기

1. 특정 시점을 갖는 날짜 객체 생성

```python
someday = dt.datetime(2018, 8, 30, 13, 26, 55)
day_str = someday.strftime("%y-%m-%d %H:%M:%S")
print(day_str)
```

2. 존재하지 않는 시점으로 설정할 경우 에러

```python
someday = dt.datetime(2018, 12, 32, 0, 0, 0)
day_str = someday.strftime("%y-%m-%d %H:%M:%S")
print(day_str)
```

3. 날짜 객체의 특정 성분 변경하기

```python
foo = dt.datetime.now()
print(foo.strftime("%y-%m-%d %H:%M:%S"))

change_date = foo.replace(year=2018, day=16, hour=15)
print(change_date.strftime("%y-%m-%d %H:%M:%S"))
```

### #04 날짜 계산

1. 두 날짜 간의 차이 계산하기

```python
dt1 = dt.datetime.now()                             # 오늘 날짜 만들기
dt2 = dt.datetime(dt1.year+1, 1, 1, 0, 0, 0)        # 내년 1월 1일 자정

td = dt2 - dt1          # 두 날짜의 차를 구한 결과는 timedelta 객체형식이 됨
print(td)
```

2. timedelta 객체의 프로퍼티

```python
print(td.days)          # 날짜만 추출

print(td.seconds)       # 날짜를 제외하고 시간, 분, 초 단위를 모두 초로 합산한 값
print("올해는 %d일 남았습니다." % td.days)
```

3. timedelta 객체의 메서드

```python
result = td.total_seconds()
print(result)
```

4. 며칠 후의 시각 계산하기

```python
d = dt.timedelta(days=100, seconds=3600)

now_time = dt.datetime.now()

after_time = now_time + d

print(after_time.strftime("%Y-%m-%d %H:%M:%S"))
```
