### #02 모듈에 별칭 적용하기

1. 함수 모듈에 별칭 적용하기

```python
from mylibrary import MyMod1 as hello

# 모듈에 포함된 기능들은 `별칭이름.함수명` 형식으로 접근
print(hello.plus(3, 4))
print(hello.minus(3, 4))
```

2. 클래스 모듈에 별칭 적용하기

```python
# MyMod2 모듈에 User라는 별칭을 적용해 import
from mylibrary import MyMod2 as User

mem = User.Member('파이썬학생', 'leekh4232@gmail.com')
mem.view_info()
```

3. 객체 모듈에 별칭 적용하기

```python
from mylibrary import MyMod3 as Myobject

print(Myobject.my_calc.plus(30, 15))
print(Myobject.my_calc.minus(30, 15))
```

### #03 모듈 내의 기능 직접 참조

1. 모듈 내의 함수를 직접 참조하기

```python
from mylibrary.MyMod1 import plus

# 골라서 가져온 기능들은 앞에 모듈 이름을 적용하지 않고 직접 접근한다
print(plus(3, 4))

# minus 함수는 import하지 않았으므로 사용 불가 --> 에러
```

2. 모듈 내의 클래스를 직접 참조하기

```python
# MyMod2 모듈에서 Member라는 이름의 기능만 지정해서 참조
from mylibrary.MyMod2 import Member

# from 절로 특정 기능만 골라서 참조한 경우 모듈이름 없이 접근
# 참조된 기능이 클래스이므로 객체로 생성해야만 사용가능
mem = Member('파이썬학생', 'leekh4232@gmail.com')
mem.view_info()
```

3. 모듈 내의 객체를 직접 참조하기

```python
# 모듈에 정의된 특정 기능(여기서는 객체)만 골라서 참조
from mylibrary.MyMod3 import my_calc

print(my_calc.plus(30, 15))
print(my_calc.minus(30, 15))
```
