### #01 메서드를 통한 멤버변수의 데이터 설정 및 출력

1. 멤버변수의 값을 수정하기 위한 메서드를 갖는 클래스

```python
class Zealot:
    hp = None
    dps = None

    def state(self):
        print("상태확인 >>> hp: {0}, dps: {1}".format(self.hp, self.dps))

    def level_up(self):
        self.hp += 10
        self.dps += 2
        print("레벨업을 하여 체력(+10), 공격력(+2)이 증가합니다.")

    def attacked(self, enemy_attack):
        self.hp -= enemy_attack
        print("{0}의 공격을 받아 hp가 {1}만큼 감소합니다.".format(enemy_attack, self.hp))
```

2. 메서드 실행에 따른 멤버변수의 변화 확인

```python
zealot1 = Zealot()
zealot1.hp = 100
zealot1.dps = 20

zealot1.state()
zealot1.attacked(30)
zealot1.state()
zealot1.level_up()
zealot1.state()
```

### #02 생성자를 갖는 클래스 사용하기

1. 클래스 정의

```python
# 마린 클래스 정의
class Marine:
    name = None
    hp = None

    # 파라미터를 활용해 객체의 속성을 초기화하는 생성자
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp
        print("[{0}] You wanna piece of me, boy?".format(self.name))

    # 객체의 속성을 일괄 출력하는 메서드
    def state(self):
        print("[{0}] hp: {1}".format(self.name, self.hp))

    # 레벨업 --> 모든 속성(멤버변수)들이 증가한다
    def level_up(self):
        self.hp += 5
        print("[{0}] 레벨업을 하여 체력(+5)이 증가합니다.".format(self.name))

    # 공격을 받음 --> 상대의 공격력만큼 hp가 감소한다
    def attacked(self, enemy_attack):
        self.hp -= enemy_attack
        print("[{0}] 적의 공격을 받아 hp가 {1} 감소합니다.".format(self.name, enemy_attack))
```

2. 생성자를 통한 객체 속성 초기화

```python
m1 = Marine("마린1", 100)
m2 = Marine("마린2", 100)

m1.state()
m2.state()

m1.attacked(30)
m2.attacked(20)

m1.state()
m2.state()

m1.level_up()
m2.level_up()

m1.state()
m2.state()
```
