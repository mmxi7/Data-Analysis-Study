## 2. 인덱싱과 슬라이싱

### #01 리스트의 인덱싱

```python
# 정방향 혹은 역방향 인덱스를 사용한 리스트 원소 접근

mylist = [10, 20, 30]           # 실습을 위한 샘플 데이터
print(mylist[1])                # 1번째 원소에 접근 -> 0부터 카운트
print(mylist[-1])               # -1번째 원소에 접근 -> -1부터 카운트
```

```python
first = mylist[0]               # 0번째 리스트의 원소를 다른 변수에 복사
print(first)
```

```python
mylist[1] = 200                 # 1번째 리스트의 원소를 다른 변수에 복사
print(mylist)
```

### #02 리스트의 슬라이싱

```python
# 1. 리스트의 구간 설정 방법

# 실습을 위한 샘플 데이터
mylist = [60, 70, 80, 90, 100]

print(mylist[1:3])              # list의 1번째부터 3번째 전까지 -> 3번째 원소 미포함

print(mylist[1:-2])             # list에서 1번째부터 뒤에서부터 2번째 다음까지
```

```python
# 2. n번째 원소부터 끝까지, 처음부터 n번째 원소까지, 모든 원소

print(mylist[3:])               # 3번째부터 끝까지, 3번째 원소 포함

print(mylist[:3])               # 처음부터 3번째 전까지, 3번째 원소 미포함

print(mylist[:])                # list의 모든 원소
```

### #03 슬라이싱을 활용한 리스트의 원소 교체 및 삭제

```python
# 1. 슬라이싱으로 지정된 원소의 수와 같은 길이를 갖는 리스트의 대입

foo = [1, 2, 3, 4, 5]

# 리스트의 1번째부터 4번째 전까지 3개의 원소를 다른 값으로 일괄 교체
foo[1:4] = [20, 30, 40]
print(foo)
```

```python
# 2. 슬라이싱으로 지정된 범위에 빈 리스트를 대입

bar = [11, 22, 33, 44, 55]

# 3개의 원소 범위에 빈 리스트 대입 --> 삭제
bar[1:4] = []
print(bar)
```

### #04 리스트의 확장, 축소


```python
# 1. 슬라이싱으로 지정된 범위에 더 긴 길이를 갖는 리스트 대입

hello = [6, 7, 8, 9, 10]

# 3개의 원소 범위에 7개의 원소를 갖는 리스트 대입 --> 확장
hello[1:4] = [100, 200, 300, 400, 500, 600, 700]
print(hello)
```

```python
# 2. 슬라이싱으로 지정된 범위에 더 짧은 길이를 갖는 리스트 대입

world = [11, 12, 13, 14, 15]

# 3개의 원소 범위에 2개의 원소를 갖는 리스트 대입 --> 축소
world[1:4] = [1000, 2000]
print(world)
```

## 연습문제

```python
# 리스트 numbers와 문자열 direction이 매개변수로 주어지는 solution 함수가 있습니다.
# 리스트 numbers의 원소를 direction 방향으로 한 칸씩 회전시킨 리스트를 return하도록 solution 함수를 완성하세요.

def solution(numbers, direction):
    answer = []                 # 회전된 리스트를 저장할 변수

    if direction == "right":
        answer = numbers[-1:] + numbers[0:-1]


    if direction == "left":
        answer = numbers[1:] + numbers[0:1]


    return answer               # 회전된 리스트 반환

print(solution([1, 2, 3], "right"))
print(solution([4, 455, 6, 4, -1, 45, 6], "left"))
```
