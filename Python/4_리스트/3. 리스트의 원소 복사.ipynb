## 3. 리스트의 원소 복사

### #01 일반 변수의 복사

```python
# 1. 일반 변수끼리의 대입

a = 100
b = a
print("a =", a, "; b =", b)
```

```python
# 복사본을 변경하더라도 원본에는 변화가 없다. (반대의 경우도 마찬가지)

b = 200
print("a =", a, "; b =", b)
```

### #02 연속성 자료형의 복사

```python
# 1. 리스트, 튜플 등의 연속성 자료형에 대한 복사

foo = [1, 2, 3]
bar = foo
print("foo =", foo, "; bar", bar)

# -> 참조된 복사본을 변경하면 원본도 함께 변경됨. (반대의 경우도 마찬가지)
bar[1] = 20
print("foo =", foo, "; bar =", bar)
```

### #03 연속성 자료형의 값 복사 방법

```python
# 1. 고전적 방법 : 각 원소의 개별 복사

a = [1, 2, 3]
b = [0, 0, 0]

# 각 원소를 하나씩 개별 복사함
b[0] = a[0]
b[1] = a[1]
b[2] = a[2]
print("a =", a, "; b =", b)

# -> 복사본을 변경하더라도 원본에 변화 없음
b[2] = 1000
print("a =", a, "; b =", b)
```

```python
# 2. 슬라이싱을 활용한 값 복사

# 슬라이싱을 활용한 값 복사
hello = [100, 200, 300]
cp2 = hello[:]

# -> 복사본을 변경하더라도 원본에 영향 없음
cp2[1] = 12345
print(hello)
print(cp2)
```

```python
# 3. 리스트 객체의 copy() 메서드(=함수)를 활용한 리스트의 복사

# 리스트 객체의 함수를 사용하는 방법
world = ["hello", "python", "world"]
cp3 = world.copy()

# -> 복사본을 변경해도 원본에 영향 없음
cp3[0] = "nice"
print(world)
print(cp3)
```

### #05 리스트, 튜플의 구조분해

```python
# 1. 원소의 1:1 복사

# 리스트의 구조분해
a, b = ["python", "bigdata"]
print(a)
print(b)
```

```python
# 2. 구조분해의 에러 케이스

# 선언된 변수의 수와 리스트의 원소 수가 다르면 에러 발생
a, b, c = [10, 20, 30, 40]
```

```python
# 3. 함수 리턴 값의 구조분해

# 두 수의 나눗셈 결과를 몫과 나머지로 반환하는 함수
def divmod(a, b):
    # 몫과 나머지를 리스트로 묶어서 반환
    return [a // b, a % b]

# 함수 호출 --> 리턴 결과는 리스트
result = divmod(7, 3)
print(result)

# 리턴 결과를 구조분해로 받기
x, y = divmod(7, 3)
print("몫: ", x, ", 나머지: ", y)
```
