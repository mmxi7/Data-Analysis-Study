```python
### #02 리스트에 포함되어 있는지를 확인하는 IN, NOT IN 연산자
```

```python
# 1. in 연산자와 not in 연산자 사용하기

mylist = [5, 3, 7, 1, 9]
target = 7

# target이 mylist에 포함되어 있는지 확인
# <-- 포함되어있으면 True
find = target in mylist
print(find)

# target이 mtlist에 포함되어 있지 않은지 확인
# <-- 포함되지 않으면 True
not_find = target not in mylist
print(not_find)
```

```python
# 2. 조건문에서 in 연산자를 사용하여 새로운 아이디가 기존 회원 목록과 중복되는지 확인

user_id_list = ["mike", "tom", "jane"]

user_id = input("아이디를 입력하세요: ")            # 사용자로부터 입력받은 아이디가

if user_id in user_id_list:                        # 이미 사용 중인지 확인
    print("죄송합니다. 해당 아이디는 사용 중입니다.")
else:
    print("사용 가능한 아이디입니다.")
```

```python
# 3. 조건문에서 not in 연산자를 사용해 새로 가입하는 아이디가 기존 회원 목록과 중복되는지 확인

user_id_list = ["mike", "tom", "jane"]

user_id = input("아이디를 입력하세요: ")            # 사용자로부터 입력받은 아이디가

if user_id not in user_id_list:                        # 이미 사용 중인지 확인
    print("사용 가능한 아이디입니다.")
else:
    print("죄송합니다. 해당 아이디는 사용 중입니다.")
```

```python
### #04 WHILE문을 사용한 리스트의 원소 탐색
```

```python
# 1. 반복문의 특성과 리스트 인덱스의 특성을 일치시킨다.

mylist = [10, 20, 30, 40, 50]
size = len(mylist)              # 리스트의 전체 원소 수
i = 0                           # 0부터 카운트 할 리스트의 인덱스

# 인덱스가 전체 원소 수보다 작은 동안 반복
while i < size:                 # 반복 범위: 0~4
    print(mylist[i])
    i += 1                      # 다음 원소 탐색을 위해 인덱스 1 증가
```

```python
# 2. 리스트의 모든 원소를 탐색해 합계 구하기

성적표 = [92, 88, 76, 90, 82]
과목수 = len(성적표)
i = 0
합계점수 = 0

# 인덱스가 과목수보다 작은 동안 반복
while i < 과목수:
    합계점수 += 성적표[i]
    i += 1

평균점수 = 합계점수 / 과목수

print("총점: ", 합계점수, "/ 평균: ", 평균점수)
```

```python
# 정수가 담긴 배열 array와 정수 n이 매개변수로 주어질 때,
# array에 n이 몇 개 있는지를 return하도록 solution함수를 완성해 보세요.

def solution(array, n):
    answer = 0                  # n이 등장하는 횟수를 저장할 변수

    for m in array:
        if n == m:
            answer += 1

    return answer

print(solution([1, 1, 2, 3, 4, 5], 1))
print(solution([0, 2, 3, 4], 1))
```
