```python
# 1. 리스트간의 덧셈

# 실습을 위한 리스트 정의
list1 = [10, 20, 30]
list2 = [40, 50, 60]
print("list1:", list1)
print("list2:", list2)

# 리스트끼리의 덧셈은 단순 병합
list3 = list1 + list2
print("list1 + list2:", list3)
```

```python
# 아이스 아메리카노는 한잔에 5500원입니다.
# 머쓱이가 가지고 있는 돈 money가 매개변수로 주어질 때, 머쓱이가 최대로 마실 수 있는 아메리카노의 잔 수와 남는 돈을 순서대로 담은 리스트를 return하도록 solution 함수를 완성해 보세요.

def solution(money):
    items = [
        money // 5500,       # 최대로 마실 수 있는 아메리카노의 잔 수
        money % 5500        # 남는 돈
    ]

    return(items)

print(solution(15000))
```
