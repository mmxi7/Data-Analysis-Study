## 5. 리스트의 탐색(2)

### 01 FOR문을 사용한 리스트의 원소 탐색

```python
# 1. 리스트의 인덱스 번호에 대한 반복 범위 설정

# 원소의 수를 조회한 다음 직접 반복범위 지정하기
# -> 단점: 각 원소가 저장되는 변수가 없으므로 원소에 대한 접근을 리스트를 통해 직접 수행해야 함

seasons = ['봄', '여름', '가을', '겨울']

size = len(seasons)
for i in range(0, size):
    print(i, "번째 원소: ", seasons[i])
```

```python
# 2. 리스트 자체를 탐색하는 for문

msg = ['python', 'is', 'good']

for i in msg:
    print(i)
```

```python
# 3. 리스트 원소의 총점, 평균 구하기

score = [100, 82, 98, 76, 51]

total = 0               # 총점을 저장할 변수

for v in score:
    total += v

avg = total // len(score)

print("총점: ", total, ", 평균: ", avg)
```

```python
# 4. 리스트의 인덱스 번호와 원소에 대한 동시 탐색

city = ['서울', '인천', '대구', '대전', '광주', '부산']

for i, v in enumerate(city):
    print(i, "번째 값:", v)
```

### #02 2차원 리스트 탐색 패턴(1)

```python
# STEP 1: 2차원 리스트의 원소 출력하기


mydata = [
    [88, 72, 83, 90],
    [63, 77, 72, 80]
]

size = len(mydata)                                      # 2차원 리스트의 길이는 행의 수를 반환
print("size =", size)

for i in range(0, size):                                # 행의 수만큼 반복하는 구성
    print(i, "번째 데이터 ->", mydata[i])                # i번째 행에 포함되어있는 개별 리스트를 출력



# STEP 2: 2차원 리스트의 각 원소에 대한 반복문 구성


    total = 0                                           # 각 행 단위 합계이므로 매 행마다 합계 변수를 새로 초기화
    size = len(mydata)                                  # 각 행에 포함된 개별 리스트의 길이

    for j in range(0, size):                            # mydata[i]에 대한 새로운 반복문 구성
        total += mydata[i][j]

    average = total / size                              # 행 단위 평균값

    print(i, "번째 데이터 ==> 총합 =", total, "평균 =", average)
```

### #03 2차원 리스트 탐색(2)

```python
# STEP 1: 리스트를 반복문에 직접 적용한 경우


mydata = [
    [88, 72, 83, 90],
    [63, 77, 72, 80]
]

for row in mydata:                                        # 행의 수만큼 반복하는 구성
    print("이번에 출력되는 데이터 ->", row)                 # mydata의 원소 중 하나인 리스트를 출력, 몇 번째 항목인지는 알 수 없음


# STEP 2: 지식 반복문 구성


    total = 0                                             # 각 행 단위 합계이므로 매 행마다 합계 변수를 새로 초기화

    for col in row:                                       # row에 대한 새로운 반복문 구성
        total += col

    average = total / len(row)                            # 행단위 평균값

    print(row, "에 대한 총합 =", total, ", 평균 =", average)
```

## 연습문제

머쓱이는 학교에서 키 순으로 줄을 설 때 몇 번째로 서야 하는지 궁금해졌습니다.  
머쓱이네 반 친구들의 키가 담긴 정수 배열 array와 머쓱이의 키 height가 매개변수로 주어질 때,  
머쓱이보다 키 큰 사람 수를 return하도록 solution 함수를 완성해 보세요.

```python
def solution(array, height):
    answer = 0              # 키가 더 큰 사람 수를 저장할 변수

    for i in array:
        if i > height:
            answer += 1

    return answer           # 키가 더 큰 사람 수 반환

print(solution([149, 180, 192, 170], 167))
print(solution([180, 120, 140], 190))
```
