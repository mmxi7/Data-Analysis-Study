### #01 리스트와 관련된 내장 함수

#### 1. `len()` : 리스트 길이 조회

```python
mylist = [1, 2, 3]
print(len(mylist))
```

#### 2. `del()` : 리스트의 원소 삭제

```python
mylist = [1, 2, 3]
del(mylist[1])
print(mylist)
```

#### 3. `sorted()` : 리스트 정렬

```python
mylist = [4, 2, 5, 3, 1]            # 원본 리스트
print(mylist)

asc = sorted(mylist)                # 원소 순차 정렬
print(asc)

print(mylist)                       # 원본 확인
```

#### 4. 리스트의 역순 정렬

```python
mylist = [10, 12, 11, 14, 13]
desc = sorted(mylist, reverse = True)
print("역순정렬:", desc)
print("원본:", mylist)
```

#### 5. 최대, 최소, 합계 구하기

```python
a = [1, 2, 3]
print(max(a))
print(min(a))
print(sum(a))
```

#### 6. 평균 구하기

```python
avg = sum(a) / len(a)
print(avg)
```

### #02 리스트와 튜플의 상호 변환

#### 1. `list()` 함수를 사용해 리스트로 변환

```python
# 문자열 > 글자들의 모음
a = "Python"
b = list(a)
print(b)

# 튜플 > 리스트
c = (1, 2, 3)
d = list(c)
print(d)
```

#### 2. `list()` 함수를 사용한 복사본 리스트 생성

```python
# 파라미터가 리스트인 경우, 원본 리스트를 복사하여 새로운 리스트를 생성
e = [10, 20, 30]
f = list(e)
print(f)
```

#### 3. `tuple()` 함수를 사용해 튜플로 변환

```python
# 문자열 > 글자들의 모음
a = "Python"
b = tuple(a)
print(b)

# 리스트 > 튜플
c = [1, 2, 3]
d = tuple(c)
print(d)
```

#### 4. `tuple()` 함수를 사용한 복사본 튜플 생성

```python
# 파라미터가 튜플인 경우, 원본 튜플을 복사해 새로운 튜플 생성
e = (10, 20, 30)
f = tuple(e)
print(f)
```

### #04 리스트 객체의 메서드

#### 1. 특정 원소의 수 카운트하기

```python
# 주어진 값과 일치하는 원소의 개수 세기
mylist = [1, 2, 3, 2, 4, 2]
c = mylist.count(2)
print(c)
```

#### 2. 특정 원소가 처음으로 나타나는 위치 찾기

```python
mylist = [1, 2, 3, 10, 1, 2, 3, 10]
x = mylist.index(10)
print(x)
```

#### 3. 원소를 찾지 못할 경우 에러 발생

```python
y = mylist.index(5)
```

#### 4. 리스트의 맨 뒤에 원소 추가하기

```python
# 맨 뒤에 내용 추가 >> 원본에 직접 반영
mylist = [1, 2, 3]
mylist.append(4)
print(mylist)

# 리스트를 append 할 경우 2차 리스트로 삽입
mylist = [1, 2, 3]
mylist.append([5, 6])
print(mylist)
```

#### 5. 리스트의 중간에 원소 삽입하기

```python
mylist = [1, 2, 3]
print(mylist)

# 1번째 위치에 10을 삽입, 기존의 항목들은 뒤로 밀려난다
mylist.insert(1, 10)
print(mylist)
```

#### 6. 리스트의 원소 추출

```python
# 맨 마지막 요소를 꺼내 리턴하고 원본에서는 삭제
mylist = [1, 2, 3]
k = mylist.pop()
print(k)
print(mylist)
```

#### 7. 리스트의 확장과 축소

```python
# extend() 메서드를 호출한 원본에 파라미터로 전달된 리스트를 덧붙인다
mylist = [1, 2, 3]
addon = [10, 9, 8, 7]
mylist.extend(addon)
print(mylist)

# 주어진 값과 일치하는 첫 번째 원소를 삭제
mylist = [1, 2, 3, 10, 1, 2, 3, 10]
mylist.remove(10)
print(mylist)
```

#### 8. 리스트의 역순 배치

```python
mylist = [1, 7, 5, 3, 9]
mylist.reverse()
print(mylist)
```

#### 9. 리스트 원소의 오름차순, 내림차순 정렬

```python
# 순차정렬 >> 객체 원본에 반영됨
mylist = [2, 5, 1, 4, 3]
mylist.sort()
print(mylist)

# 역순정렬 >> 객체 원본에 반영됨
mylist = [2, 5, 1, 4, 3]
mylist.sort(reverse = True)
print(mylist)
```

#### 10. 문자열을 잘라 리스트로 변환

```python
# 문자열을 주어진 글자를 기준으로 잘라서 리스트로 변환
# -> 문자열 객체에 포함된 함수 사용
text = "Hello,Python,World,Good"
mylist = text.split(",")
print(mylist)
```

#### 11. 리스트를 문자열로 결합

```python
# 문자열($)이 파라미터로 전달된 리스트의 각 원소를 연결하는 기호로 사용됨
dl = "$"
mylist = ['Hello', 'Python', 'World', 'Good']
text = dl.join(mylist)
print(text)

# 별다른 기호 없이 리스트를 하나의 문자열로 연결하기
empty = ""
mylist = ['Hello', 'Python', 'World', 'Good']
text = empty.join(mylist)
print(text)
```

#### 12. 공백으로 구분된 하나의 문자열로 결합

```python
space = " "
mylist = ['Hello', 'Python', 'World', 'Good']
text = space.join(mylist)
print(text)
```
