## 3. 문자열 관련 함수

### 01 문자열과 관련된 내장 함수

1. `len()`: **문자열 길이 조회**

```python
# 전체 글자수 세기 -> 한글, 영문 구분 없이 1글자씩 카운트
mystr = "Hello Python"
length = len(mystr)
print(length)

# 반환되는 값으 변수에 저장하지 않고 직접 출력도 가능
print(len("안녕하세요 파이썬"))
```

2. `str()`: 다른 타입의 변수를 문자열로 변환

```python
# 정수 -> 문자열
num1 = 123
num1_s = str(num1)
print(type(num1_s))
print(num1_s)
```

```python
# 불리언 -> 문자열
bool1 = True
bool1_s = str(bool1)
print(type(bool1_s))
print(bool1_s)
```

```python
# 실수 -> 문자열
num2 = 123.45
num2_s = str(num2)
print(type(num2_s))
print(num2_s)
```

### #02 문자열 객체의 메서드

1. `count()`: 문자열 안에 포함된 특정 글자나 단어의 개수 세기

```python
mystr = "파이썬은 웹 프로그래밍이나 데이터 분석에 많이 사용되는 언어입니다."

# 글자 카운트
print(mystr.count("이"))
print(mystr.count("가"))

# 단어 카운트
print(mystr.count("파이썬"))
print(mystr.count("Python"))
```

2. `find()`: 특정 글자나 단어가 처음 나타나는 위치 조회 (찾지 못할 경우 -1 리턴)

```python
message = "내가 좋아하는 것에만 집중하기로 했다."

# 글자의 위치(인덱스) 조회
print(message.find("가"))
print(message.find("나"))

# 단어의 시작 위치 조회
print(message.find("집중"))
print(message.find("파이썬"))
```

3. `rfind()`: 특정 단어나 글자가 마지막으로 나타나는 위치 조회 (찾지 못할 경우 -1 리턴)

```python
sentence = "사람의 사람에 의한 사람을 위한"

# 글자 검색
print(sentence.rfind("사"))
print(sentence.rfind("자"))

# 단어 검색
print(sentence.rfind("사람"))
print(sentence.rfind("인간"))
```

4. `isupper()`: 대문자만으로 구성되었는지 검사하기

```python
# 문자열이 대문자로 구성된 경우 True
test3 = "HELLO"
print(test3.isupper())

# 문자열이 대문자로 구성되지 않은 경우 False
test4 = "Hello"
print(test4.isupper())
```

5. `islower()`: 소문자만으로 구성되었는지 검사하기

```python
# 문자열이 소문자로 구성된 경우 True
test5 = "hello"
print(test5.islower())

# 문자열이 소문자로 구성되지 않은 경우 False
test6 = "Hello"
print(test6.islower())
```

6. upper(), lower(): 대소문자 변환

```python
mystr = "Hello Python"

up = mystr.upper()          # 모든 글자를 대문자로
print(up)

lw = mystr.lower()          # 모든 글자를 소문자로
print(lw)

print(mystr)                # 원본 문자열은 변화 없음
```

7.
- swapcase(): 대소문자 상호 변환
- capitalize(): 문장의 첫 글자만 대문자 변환(나머지는 소문자 변환)
- title(): 각 단어의 첫 글자만 대문자 변환

```python
mystr = "python is VERY EASY"
print(mystr.swapcase())

print(mystr.capitalize())

print(mystr.title())
```

10. strip(), lstrip(), rstrip(): 공백 제거

```python
k = "      python      "
print("[" + k + "]")

x = k.strip()               # 앞뒤 공백 제거
print("[" + x + "]")

y = k.lstrip()              # 왼쪽 공백 제거
print("[" + y + "]")

z = k.rstrip()              # 오른쪽 공백 제거
print("[" + z + "]")
```

11. 문자열 바꾸기 -> replace("원본값", "변경할 값")

```python
myMessage = "Hello World"

# 변수.replace(A, B), 변수에서 A를 B로 변경한 결과를 반환
newMsg = myMessage.replace("World", "Python")
print(newMsg)

print(myMessage)            # 원본은 변화 없음
```

12. 문자열 자르기 -> 문자열객체.split(기준값)

```python
# 문자열을 잘라서 리스트로 변환하기
sample1 = "파이썬,데이터분석,머신러닝"
slist1 = sample1.split(",")
print(slist1)

# split() 메서드에 파라미터가 없을 경우 공백을 기준으로 잘라냄
sample2 = "파이썬 데이터분석 머신러닝"
slist2 = sample2.split()
print(slist2)
```

13. 숫자로만 구성된 문자열인지 검사 -> 문자열객체.isnumeric()

```python
# 숫자로만 구성된 문자열인 경우 True
test1 = "123"
print(test1.isnumeric())

# 숫자로만 구성되지 않은 문자열인 경우 False
test2 = "Hello123"
print(test2.isnumeric())
```

## 연습문제

1. 문자열 배열 strlist가 매개변수로 주어집니다.
strlist 각 원소의 길이를 담은 배열을 return하도록 solution 함수를 완성해주세요.

```python
def solution(strlist):
    answer = []
    for s in strlist:
        answer.append(len(s))

    return answer

print(solution(["We", "are", "the", "World!"]))
print(solution(["I", "Love", "Programmers."]))
    
```

2. 영어 대소문자로 이루어진 문자열 my_string이 매개변수로 주어질 때,
my_string을 모두 소문자로 바꾸고 알파벳 순서대로 정렬한 문자열을 return하도록 solution 함수를 완성해보세요.

```python
def solution(my_string):
    mylist = list(my_string.lower())
    answer = "".join(sorted(mylist))
    return answer

print(solution("Bcad"))
print(solution("heLLo"))
print(solution("Python"))
```
