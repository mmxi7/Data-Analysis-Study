## 1. 예외처리의 이해

### #01 자주 겪을 수 있는 에러 케이스

1. 잘못된 연산 에러 (어떤 수를 0으로 나누는 경우)

```python
x = int(input("숫자를 입력하세요."))

# x 에 0 이 입력된 경우 에러 발생 -> ZeroDivisionError
y = 100 / x
print("결과는 %d입니다." % y)
```

2. 리스트의 인덱스 접근 에러

```python
mylist = [92, 76, 83, 64, 95]
mysum = 0

x = int(input("숫자를 입력하세요."))

# 0 부터 입력값 x 전까지 반복
for i in range(x):
    mysum += mylist[i]

print("합계는 %d입니다." % mysum)
```

### #02 고전적인 예외 처리

1. 연산을 수행할 수 없는 경우를 IF 문으로 대비해야 함

```python
x = int(input("숫자를 입력하세요."))

# 입력값이 0인 경우에 대비해야 함
if x == 0:
    print("0으로 나눌 수 없습니다.")
else:
    y = 100 / x
    print("결과는 %d입니다." % y)
```

2. 입력값이 범위를 초과하는지 미리 검사해야 함

```python
mylist = [92, 76, 83, 64, 95]
mysum = 0

x = int(input("숫자를 입력하세요."))

# 입력값이 리스트의 길이보다 큰 경우에 대비해야 함
if len(mylist) < x:
    print("인원수가 너무 큽니다.")
else:
    # 0 부터 입력값 x 전까지 반복
    for i in range(x):
        mysum += mylist[i]

    print("합계는 %d입니다." % mysum)
```

### #03 `try~except`문 기반 예외처리

1. 기본 문법 구조

```python
try:
    x = int(input("숫자를 입력하세요."))
    y = 100 / x
    print("결과는 %d입니다." % y)
except:
    print("에러가 발생했습니다.")
```

2. 예외의 종류에 따라 다르게 대응하기

```python
try:
    x = int(input("숫자를 입력하세요."))
    y = 100 / x
    print("결과는 %d입니다." % y)
except ValueError:
    print("숫자만 입력하세요.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
except Exception:
    print("알 수 없는 에러가 발생했습니다.")
```

3. 에러 메시지를 변수에 저장하여 출력하기

```python
try:
    x = int(input("숫자를 입력하세요."))
    y = 100 / x
    print("결과는 %d입니다." % y)
except ValueError as verr:
    print("숫자만 입력하세요.", verr)
except ZeroDivisionError as zerr:
    print("0으로 나눌 수 없습니다.", zerr)
except Exception as err:
    print("알 수 없는 에러가 발생했습니다.", err)
```

4. 에러 발생 여부에 상관 없이 무조건 실행되는 블록

```python
try:
    x = int(input("숫자를 입력하세요."))
    y = 100 / x
    print("결과는 %d입니다." % y)
except ValueError as verr:
    print("숫자만 입력하세요.", verr)
except ZeroDivisionError as zerr:
    print("0으로 나눌 수 없습니다.", zerr)
except Exception as err:
    print("알 수 없는 에러가 발생했습니다.", err)
finally:
    print("처리가 완료되었습니다.")
    
```

### #04 가장 편한 실행 형태

1. 예외 상황을 통합해 대응할 수 있는 Exception 블록만 사용

```python
try:
    x = int(input("숫자를 입력하세요."))
    y = 100 / x
    print("결과는 %d입니다." % y)
except Exception as e:
    print("에러가 발생했습니다.", e)
finally:
    print("처리가 완료되었습니다.")
```

2. 리스트 탐색에 대한 예외처리

```python
mylist = [92, 76, 83, 64, 95]
mysum = 0

try:
    x = int(input("인원수를 입력하세요."))

    # 0부터 입력값 x 전까지 반복
    for i in range(x):
        mysum += mylist[i]

    print("합계는 %d입니다." % mysum)
except Exception as e:
    print("에러가 발생했습니다.", e)
finally:
    print("처리가 완료되었습니다.")
```
