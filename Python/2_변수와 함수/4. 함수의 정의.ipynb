### #05 기본값을 갖는 파라미터

```python
def sum1(x, y, z = 0):
    answer = x + y + z
    print(answer)
```

```python
# 모든 파라미터값을 전달하며 함수 호출
sum1(1, 2, 3)
```

```python
# 초기값이 설정된 파라미터는 호출 시 생략 가능 --> 생략된 파라미터는 초기값으로 설정됨
sum1(10, 20)
```

### #06 파라미터 이름 지정하기

```python
def sum2(a, b, c = 0, d = 0):
    answer = a + b + c + d
    print(answer)
```

```python
# 파라미터의 이름을 지정한 함수 호출 --> 파라미터 전달 순서 무시 가능
sum2(d = 3, c = 4, b = 2, a = 1)
```

```python
# 함수에 정의된 기본값 활용
sum2(1, b = 2)
```

### #07 값을 리턴하는 함수

```python
def plus(x, y):
    z = x + y
    return z    # 자신이 호출된 위치로 값을 돌려줌
```

```python
a = plus(10, 20)    # 리턴값을 다른 변수에 할당하기
print(a)
```

```python
print( plus(100, 200) )     # 리턴값을 갖는 함수는 그 결과를 직접 출력할 수 있다
```

```python
b = plus(5, 7) + 100    # 리턴값을 갖는 함수는 그 결과를 다른 연산에 활용할 수 있다
print(b)
```

### #08 함수의 실행 중단

```python
# return문을 중간에 만나는 경우 그 즉시 수행을 종료한다
# 리턴문 이후의 코드는 실행되지 않는다
def foo(x, y):
    return 0
    z = x + y
    return z
```

```python
# z값을 계산하기 전에 0을 리턴하므로 결과값이 0이 된다
print( foo(100, 200) )
```

### #09 함수에서의 변수의 유효성 범위

```python
a = 100
b = 200

def foo():
    k = a + b   # 함수 안에서는 함수 밖의 변수 식별 가능
    print(k)
```

```python
foo()
```

```python
print(k)    # 함수 밖에서는 함수 내부에서 생성된 변수 식별 불가
```
