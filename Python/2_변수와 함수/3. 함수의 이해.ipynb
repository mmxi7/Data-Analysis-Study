## 3. 내장함수

    파이썬에서 제공하는 미리 정의되어있는 함수

### #01 type() 내장함수 사용하기

```python
# 정수형 변수의 데이터 타입 확인
age = 20
typeOfAge = type(age)
print(typeOfAge)
```

```python
# 실수형 변수의 데이터 타입 확인
pi = 3.14
typeOfPi = type(pi)
print(typeOfPi)
```

```python
# 문자열 변수의 데이터 타입 확인
message = "Hello Python"
typeOfMessage = type(message)
print(typeOfMessage)
```

```python
# 논리형 변수의 데이터 타입 확인
ok = True
typeOfOk = type(ok)
print(typeOfOk)
```

### #02 여러 개의 값을 한 번에 출력하기

### #03 사용자 입력값 받기

```python
# 문자열 입력받기
msg1 = input("메시지를 입력하세요")
print(msg1)

# 입력 받은 내용의 데이터 타입 확인 -> 문자열(str)
t = type(msg1)
print(t)
```

### #04 문자열의 정수, 실수 변환

```python
# 입력받은 값을 정수 타입으로 변환
msg2 = input("정수를 입력하세요")
num = int(msg2)
print(num)

# 변환 결과의 데이터 타입 확인 -->  정수(int)
num_type = type(num)
print(num_type)
```

```python
# 입력받은 값을 실수 타입으로 변환
msg3 = input("실수를 입력하세요")
fnum = float(msg3)
print(fnum)

# 변환 결과의 데이터 타입 확인 --> 실수(float)
fnum_type = type(fnum)
print(fnum_type)
```

### #05 정수와 실수의 상호 변환

```python
# 정수를 실수로 변환 --> 정수 뒤에 ".0"이 추가됨
num1 = 100
num2 = float(num1)
print(num2)

# 변환 결과의 데이터 타입 확인 --> 실수(float)
t1 = type(num2)
print(t1)
```

```python
# 실수를 정수로 변환 --> 소수점 버림
num3 = 123.78
num4 = int(num3)
print(num4)

# 변환 결과의 데이터 타입 확인 --> 정수(int)
t2 = type(num4)
print(t2)
```

### #07 간단한 연산 기능 함수

```python
# 1. 제곱근 구하기
a = pow(2, 8)
print(a)
```

```python
# 2. 절대값 구하기
a = abs(100)
print(a)

b = abs(-100)
print(b)

c = abs(-12.345)
print(c)
```

```python
# 3. 소수점 반올림

# 소수점 반올림 후 정수만 출력
n = round(123.78)
print(n)

# 소수점 첫째 자리까지만 남기고 반올림
n2 = round(123.76, 1)
print(n2)
```
