## #01. 다양한 변수 만들기

### 1. 변수의 초기화

- 어떤 변수에 최초로 값을 할당하는 것을 **초기화**라고 한다

```python
# 변수 만들기 (오른쪽에서 왼쪽으로 대입 -> 최초대입을 초기화라고 함)
# -> 이름 = 값
a = 123
b = -456
print(a)
print(b)
```

### 2. 저장되어 있는 값의 변경

- 기존의 값이 삭제되고 새로운 값이 저장된다.

```python
# 한 번 할당된 변수는 다른 값으로 변경 가능
a = 10000
print(a)
```

### 3. 소수점을 포함하는 값을 저장하는 변수

```python
# 실수형 변수 만들기
c = 1.2
d = -3.45
print(c)
print(d)
```

### 4. 논리형 변수

- 참, 거짓을 표현할 수 있는 변수
- True와 False에 첫 글자가 대문자인 것에 주의한다.

```python
a = True
print(a)
```

```python
b = False
print(b)
```

### 5. 문자열 변수

```python
# 문자열 변수
msg = '안녕하세요. 파이썬'
print(msg)
```

### 6. 내용이 없는 변수

- 메모리 상에 공간만 점유한 상태로 데이터가 저장되어 있지는 않다.
- 추후 프로그램이 동작하는 과정에서 값을 저장하기 위해 미리 메모리 공간만 설정해 놓은 상태

```python
# 내용이 없는 변수
hello = None
print(hello)
```
