### #01 형식문자 포매팅

1. 숫자형 형식문자의 사용

```python
# "%" 연산자 뒤의 값이 "%d"를 치환한다.
str1 = "나는 %d원을 갖고 있다." % 10000
print(str1)
```

```python
# 실수값 표현하기
money = 123.456789
print("[%f]원 입니다." % money)
```

2. 문자열 형식문자의 사용

```python
lang = "Python"
str2 = "Hello %s" % lang
print(str2)
```

3. 두 개 이상의 형식문자 사용

```python
# 형식문자를 포함한 문장만 별도의 변수로 지정
str3 = "%s 날짜는 %d년 %d월 %d일 입니다."

# 여러 형식문자를 치환할 경우 값들을 괄호로 묶는다
msg1 = str3 % ("정모", 2023, 1, 1)
print(msg1)

msg2 = str3 % ("약속", 2023, 3, 5)
print(msg2)
```

4. 정수의 자릿수 표현

```python
# 자릿수 지정하기 -> 오른쪽 정렬, 빈자리는 공백으로 지정
print("[%10d]원을 입금했습니다." % 120000)

# 자릿수 지정하기 -> 오른쪽 정렬, 빈자리는 0으로 지정
print("[%010d]원을 입금했습니다." % 120000)
```

5. 문자열의 자릿수 표현

```python
# 문자열의 자릿수 지정하기 -> 빈자리는 공백지정만 가능
print("[%10s]에 살고 있습니다." % "대한민국")
```

6. 실수의 자릿수 표현

```python
myMoney = 12345.6789

# 소수점 3자리까지 표현하기 -> 소수점 4자리에서 반올림
print("[%0.3f]원 입니다." % myMoney)

# 소수점 3자리까지 표현하되 전체 글자수는 10글자(빈자리 공백)
# 소수점이 자릿수에 포함 -> 오른쪽 정렬
print("[%010.3f]원 입니다." % myMoney)
```

7. 실수의 자릿수 표현 + 빈자리 0으로 채우기

```python
# 소수점 3자리까지 표현하되 전체 글자수는 10글자 (빈자리 0)
# -> 빈 자리를 0으로 채우는 경우 오른쪽 정렬만 가능
print("[%010.3f]원 입니다." % myMoney)
```

### #02 format 메서드 포매팅

1. 인덱스 번호를 사용한 포메팅

```python
# 0번째 위치에 format() 함수에 전달된 0번째 값을 대입
msg1 = "I eat {0} apples"
print(msg1.format(3))

# 여러 개의 값을 혼용하기
msg2 = "{0}은 {1}년 {2}월 {3}일 입니다."
print(msg2.format("개강일", 2023, 1, 26))
```

2. 인덱스 이름을 사용한 포메팅

```python
msg3 = "{name}은 {yy}년 {mm}월 {dd}일 입니다."
print(msg3.format(mm = 10, name = "종강일", dd = 20, yy = 2023))
```

3. 숫자 형식과 이름 형식의 치환문자 함께 사용하기

```python
# 홉합사용 -> 숫자형식의 치환문자가 format() 메서드 안에서 먼저 위치해야 함
msg4 = "{0}은 {yy}년 {mm}월 {dd}일 입니다."
print(msg4.format("생일", yy = 2023, mm = 1, dd = 31))

# 인덱스 번호는 format() 함수 내의 순서 -> 형식 문자열 안에서는 자유롭게 배치 가능
fmt5 = "{1} {name}는 이 {0}입니다."
rst = fmt5.format("책", "좋은", name = "교과서")
print(rst)
```
