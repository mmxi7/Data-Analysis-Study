## 1. 문자열의 이해

### #01 문자열 다루기

1. 문장을 표현하는 글자들의 모임을 값으로 갖는 변수

```python
# 쌍따옴표나 홑따옴표로 감싸야 한다 >> 따옴표 혼용은 불가
msg1 = "Life is too short"
msg2 = 'You need Python'
print(msg1)
print(msg2)
```

### #02 문자열의 특성

1. 문자의 배열(리스트)

```python
mymsg = "Hello"

# 문자열을 반복문으로 탐색하면 문자 하나씩 출력됨
for m in mymsg:
    print(m)
```

2. 문자열 간의 연산 >> 덧셈만 가능

```python
# 문자열의 덧셈은 문자열을 연결하는 기능
msg1 = "Hello"
msg2 = "Python"
print(msg1 + ", " + msg2)

# 따옴표로 감싸면 무조건 문장 >> 숫자가 아니므로 연산 결과도 문자열 연산 규칙을 따름
pay = "20000"
bonus = "100"
print(pay + bonus)
```

3. 문자열의 곱셈

```python
# 문자열과 숫자의 곱셈은 동일 내용을 여러 번 반복
name = "Python"
print(name * 5)
```

### #03 이스케이프 문자의 사용

1. 따옴표 안의 따옴표

```python
# 쌍따옴표 안에 쌍따옴표 넣기 >> 이스케이프 문자 사용
case1 = "Life is \"too\" short"
print(case1)

# 홑따옴표 안에 홑따옴표 넣기 >> 이스케이프 문자 사용
case2 = 'Life is \'too\' short'
print(case2)
```

2. 줄바꿈 문자 (개행문자)

```python
# 줄바꿈 >> \n
print("Life is too short\nYou need Python")
```

3. 탭 문자

```python
# 탭키 >> \t
print("Life is too short\tYou need Python")
```

4. 순수한 역슬래시 출력

```python
# 순수한 역슬래시 출력 >> \\
print("Life is too short\\You need Python")
```

5. 문자열 안에 따옴표를 포함시키는 다른 방법

```python
str1 = "You 'need' Python"
print(str1)

str2 = 'You "need" Python'
print(str2)
```

### #04 문자열 인덱싱

```python
mystr = "Life is too short, You need Python"
print(mystr)

print(mystr[3])             # 3번째 문자 추출(0부터 카운트)

print(mystr[-3])            # 뒤에서 3번째 글자 추출(-1부터 카운트)
```

### #05 문자열의 슬라이싱

```python
mystr = "Life is too short, You need Python"
print(mystr)

print(mystr[0:4])

print(mystr[5:7])

print(mystr[19:])

print(mystr[:17])

print(mystr[:])

print(mystr[19:-7])

print(mystr[7:5])           # 인덱스 번호가 잘못되면 출력되는 내용 없음
```

## 연습문제

1. 문자열 my_string과 정수 n이 매개변수로 주어질 때, my_string에 들어있는 각 문자를 n만큼 반복한 문자열을 return하도록 solution 함수를 완성해보세요.

- mystring : Hello
- n : 3
- result : hhheeellllllooo

```python
def solution(str, n):
    answer = ""
    for m in str:
        answer += m * n

    return(answer)

print(solution("hello", 3))
```

2. 문자열 my_string이 매개변수로 주어집니다. my_string을 거꾸로 뒤집은 문자열을 return하도록 solution 함수를 완성해보세요.

- jaron > noraj
- bread > daerb

```python
def solution(my_string):
    answer = ""             # 문자를 하나씩 가져와서 쌓을 곳
    n = len(my_string)            # 문자열 길이
    for m in range(0, n):   # 문자열 길이만큼 반복
        n -= 1              # 마지막 인덱스가 4이므로 1을 빼줌
        answer += my_string[n]    # str[4]부터 누적해서 쌓아줌
    return(answer)

print(solution("jaron"))
print(solution("bread"))

# 풀이 1
def solution(my_string):
    answer = ''
    k = len(my_string)      # 문자열의 길이

    for i in range(0, k):
        # 반대쪽 인덱스를 계산
        answer += my_string[k - i - 1]

    return answer

print(solution("jaron"))
print(solution("bread"))

# 풀이 2
def solution(my_string):
    answer = ''
    k = len(my_string) # 문자열의 길이
    x = -1

    for i in range(0, k):
        # 음수 인덱스 사용
        answer += my_string[x]
        x -= 1

    return answer

print(solution("jaron"))
print(solution("bread"))

# 풀이 3
def solution(my_string):
    my_list = list(my_string)    # 문자열을 리스트로 변환
    my_list.reverse()            # 리스트 뒤집기
    
    empty = ""
    answer = empty.join(my_list) # 리스트를 문자열로 변환
    return answer

print(solution("jaron"))
print(solution("bread"))
```
