## #01 while문의 기본 구조

```python
x = 1   # 초기식
while x <= 10:
    print("x=", x)
    x += 1

print("x의 최종값 =", x)
```

```python
## #02 while문 활용하기
```

```python
# 구구단 출력하기

x = int(input("숫자를 입력하세요"))

y = 1   # 초기식
while y < 10:   # 조건식 -> 10보다 작다 이므로 10은 미포함
    z = x * y   # 반복이 진행되는 동안 수행할 명령
    print(x, "x", y, "=", z)
    y += 1  # 증감식 -> y가 1씩 증가함
```

```python
# 1~10까지의 총 합을 구하기

x = 1   # 초기식
sum = 0     # 총 합을 누적해서 저장할 변수
while x <= 10:     # 조건식 -> x가 10보다 작거나 같은 동안 반복수행
    sum += x
    print("x =", x, "-> sum =", sum)
    x += 1  # 증감식 -> 조건에 사용되는 값을 변경
```

## #03 값의 변화 단계 조절하기

```python
# 0부터 40까지 10씩 증가

a = 0
while a < 50:
    print(a)
    a += 10
```

```python
# 10부터 2까지 2씩 감소

b = 10
while b > 0:
    print(b)
    b -= 2
```

## #04 연습문제

```python
i = 0       # 1~100까지 범위를 표현하기 위한 초기식
sum = 0     # 합계를 저장하기 위한 변수 초기화

while i <= 100:
    i % 2 == 0
    sum += i * (i % 2 == 0)
    i += 1

print(sum)
```
