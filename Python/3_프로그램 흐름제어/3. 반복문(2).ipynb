## 3. 반복문(2)

### #01 for문의 기본 구조

1. 기본 for문의 예시

```python
# x가 1부터 11에 도달하기 전까지 1씩 증가하면서 반복수행

for x in range(1, 11):
    # 반복이 진행되는 동안 수행할 명령
    print("x =", x)

print("x의 최종값 =", x)
```

### #02 for문의 활용

1. 구구단 출력하기

```python
x = int(input("숫자를 입력하세요."))

# -> 1부터 10 전까지 1씩 증가하는 값을 y에 설정
for y in range(1, 10):
    # 반복이 진행되는 동안 수행할 명령
    z = x * y
    print(x, "x", y, "=", z)
```

2. 합계 구하기

```python
# 1~10까지의 총 합을 구하기
sum = 0     # 총 합을 누적해서 저장할 변수

for x in range(1, 11):      # x가 1부터 11전까지 1씩 증가하면서 반복수행
    # sum이 x만큼 스스로 증가
    # -> x가 1~10까지 변화하게 되므로 결국 sum은 1부터 10까지 순차적으로 증가
    sum += x
    print("x =", x, "-> sum =", sum)
```

### #03 값의 변화 단계 조절하기

1. 10씩 증가하는 반복문

```python
# 0부터 50 전까지 10씩 증가

for a in range(0, 50, 10):
    print(a)
```

2. 2씩 감소하는 반복문

```python
# 10부터 0 전까지 2씩 감소

for b in range(10, 0, -2):
    print(b)
```

## 연습문제

```python
# 임의의 정수 x를 입력받은 값으로 초기화하고 해당 값의 팩토리얼을 출력하는 프로그램을 구현하시오.
x = 5

facto = 1

for i in range(1, x+1):
    facto *= i

print(x, "의 팩토리얼 =", facto)
```
