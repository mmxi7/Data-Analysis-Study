## #02 조건식에 대한 다양한 유형

1. 숫자형 변수에 대한 조건 처리

```python
# money = 10000
money = 0

if money:
    print("택시를 타고")

print("집에 가자")
```

2. 문자열 변수에 대한 조건 처리

```python
# name = "Python"
name = ""

if name:
    print(name)

print("프로그래밍")
```

3. 비교식을 사용한 IF문

```python
myage = int(input('나이를 입력하세요'))

if myage == 19:
    print("19살입니다.")

if myage != 20:
    print("20살이 아닙니다.")

if myage > 19:
    print("성인입니다.")

if myage <= 19:
    print("청소년입니다.")
```

4. 논리식을 사용한 IF문

```python
point = int(input("점수를 입력하세요"))

# AND --> 전체가 참이어야 결과가 참
if point > 70 and point <= 80:
    print("C학점입니다.")

# OR --> 하나라도 참이면 전체가 참
if point <= 70 or point > 80:
    print("C학점이 아닙니다.")
```

5. 논리값에 대한 조건문

```python
# isKorean = True
isKorean = False

if isKorean == True:
    print("한국사람입니다...(1)")

if isKorean == False:
    print("한국사람이 아닙니다...(2)")

if isKorean != True:
    print("한국사람이 아닙니다...(3)")

if isKorean != False:
    print("한국사람입니다...(4)")

if isKorean:
    print("한국사람입니다...(5)")

if not isKorean:
    print("한국사람이 아닙니다...(6)")
```

## #03 else문 활용하기

1. 입력값에 따른 if~else 분기 파악하기

```python
print("시험 시작")

score = int(input("점수를 입력하세요"))

if score >= 60:
    print("합격입니다.")
else:
    print("불합격입니다.")

print("시험 끝")
```

2. boolean형 값에 대한 else문 처리

```python
answer = input("한국인입니까? (Y/N)")

if answer == "Y":
    isKorean = True
else:
    isKorean = False

# 논리값을 조건으로 사용할 경우 "만약 이 변수가 참이라면?"의 의미
if isKorean:
    print("한국사람입니다...(1)")
else:
    print("한국사람이 아닙니다...(2)")

# 논리값의 부정형을 조건으로 사용할 경우 "만약 이 변수가 거짓이라면?"의 의미
if not isKorean:
    print("한국사람이 아닙니다...(3)")
else:
    print("한국사람입니다...(4)")
```

## #04 elif를 활용한 다양한 경우의 수 나열

```python
print("시험시작")

point = int(input("점수를 입력하세요"))

if point > 90:
    print("A학점")
elif point > 80:
    print("B학점")
elif point > 70:
    print("C학점")
elif point > 60:
    print("D학점")
else: print("F학점")

print("시험 끝")
```

## #05 연습문제

### 1. 연령대 구하기

```python
age = int(input("나이를 입력하세요."))

if age > 60:
    print("당신은 노년층 입니다.")
elif age > 50:
    print("당신은 50대 입니다.")
elif age > 40:
    print("당신은 40대 입니다.")
elif age > 30:
    print("당신은 30대 입니다.")
elif age > 20:
    print("당신은 20대 입니다.")
elif age > 10:
    print("당신은 10대 입니다.")
else:
    print("당신은 어린이 입니다.")
```

### 2. 비만도 구하기

```python
Weight = int(input("몸무게를 입력하세요."))
Height = int(input("키를 입력하세요."))

if Height < 150:
    standardWeight = Height - 110
else:
    standardWeight = (Height - 110) * 0.9

BMI = (Weight - standardWeight) / standardWeight * 100

if BMI > 50:
    print("고도비만(매우위험)")
elif BMI > 30:
    print("중등도 비만(위험)")
elif BMI > 20:
    print("경도비만(주의)")
else:
    print("정상(안심)")
```
