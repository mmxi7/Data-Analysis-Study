### #02 반복문에 포함된 조건문

```python
# 1. 짝, 홀수의 합

number = int(input("숫자를 입력하세요: "))  # 입력예시: 10

# 총 합을 구하기 위해서는 반복문 전에 0으로 초기화된 변수가 필요하다
x = 0
y = 0

for i in range(1, number+1):        # -> 1~number까지 반복 수행
    if i % 2 == 0:                  # 어떤 수를 2로 나눈 나머지가 0이면 짝수
        print(i, "--> 짝수")
        x += i
    else:                           # 그렇지 않으면 홀수
        print(i, "--> 홀수")
        y += i

print("짝수들의 합:", x)
print("홀수들의 합:", y)
```

```python
# 2. 자주 사용되는 유형

for i in range(0, 5):
    print(i)
    # 조건값이 마지막이 아닌 경우만 실행하도록 if문을 구성
    if i + 1 < 5:           # i + 증가량 < 반복종료값
        print("-----")
```

```python
# 3. 1에서 100 사이의 수 중에서 3과 5의 공배수 출력하기

x = 3
y = 5

for i in range(1, 101):
    if i % x == 0 and i % y == 0:
        print(i)
```

### #03 중첩 반복문

```python
# 1. 중첩 반복문의 실행 과정 확인

for i in range(0, 3):
    print("i에 대한 반복 수행 시작 : i -->", i)

    for j in range(0, 5):
        print("::: i =", i, ", j =", j)

    print("i에 대한 반복 수행 종료")

    if i + 1 < 3:
        print("----------------------")
```

```python
# 2. 구구단

# i의 변화범위 2~9

for i in range(2, 10):

    # i가 1 증가하는 동안 j의 값이 매번 1~9까지 새로 시작
    for j in range(1, 10):
        k = i * j
        print(i, "x", j, "=", k)

    if i + 1 < 10:
        print("-----")
```

```python
# 3. 별 찍기 문제

# 행을 담당하는 반복문
for i in range(0, 5):
    # 매 행마다 출력할 내용을 초기화
    star = ""

    # 열을 담당하는 반복문 -> 행이 증가함에 따라 반복횟수가 1씩 늘어난다.
    for j in range(0, i+1):
        star += "*"

    print(star)
```

### #04 반복문의 흐름제어

```python
# 1. continue와 break의 기능 확인

y = 0
while True:
    y += 1

    if y % 2 == 0:
        continue        # 조건식으로 강제 이동

    if y > 10:
        break           # 반복문을 강제로 종료

    print("Hello Python ::", y)
```

## 연습문제

```python
# 두 개의 주사위를 던졌을 때, 눈의 합이 6이 되는 모든 경우의 수를 출력하고 경우의 수는 총 몇 가지인지 출력하시오.

count = 0           # 조건에 부합하는 경우의 수를 카운트하기 위한 변수

for i in range(1, 7):
    for j in range(1, 7):
        k = i + j

        if k == 6:
            print("i =", i, ", j =", j)
            count += 1

print("경우의 수:", count)
    
```
