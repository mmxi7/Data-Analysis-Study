## 2. 딕셔너리 관련 함수

### #01 파이썬 내장 함수

1. `len() 함수`: 딕셔너리의 원소 수 카운트

```python
point = {"korean": 98, "math": 77, "english": 82}
len(point)
```

2. `del() 함수`: 딕셔너리의 특정 원소 삭제

```python
point = {"korean": 98, "math": 77, "english": 82}
print(point)

# 특정 원소 삭제 -> 존재하지 않는 키를 삭제할 경우 에러 발생
del(point['english'])
print(point)
```

### #02 딕셔너리 객체의 메서드

1. `get()`: 딕셔너리의 값 조회

```python
dic = {"name": "철수", "phone": "010-1234-5678", "birth": "0115"}

print(dic.get("name"))              # 특정 key에 대응하는 값 얻기

print(dic.get("gender"))            # 존재하지 않는 Key에 접근하는 경우 에러가 아닌 None 반환
```

2. `get()` 메서드와 키를 통한 접근의 차이점

```python
foo = {"a": 100, "b": 200}

# get 함수는 전달하는 key가 존재하지 않을 경우 대신 반환될 값을 함께 설정 가능
print(foo.get("c", 300))
```

3. key 분리하기

```python
성적표 = {"국어": 92, "수학": 88, "영어": 72}

keys = 성적표.keys()
print(keys)

key_list = list(keys)
print(key_list)

keylist = list(성적표.keys())
print(keylist)
```

4. value 분리하기

```python
성적표 = {"국어": 92, "수학": 88, "영어": 72}

values = 성적표.values()
print(values)

value_list = list(values)
print(value_list)

valuelist = list(성적표.values())
print(valuelist)
```

5. `update()`: 딕셔너리에 다른 딕셔너리를 추가하는 메서드

```python
성적표1 = {"국어": 92, "수학": 88, "영어": 72}
성적표2 = {"물리": 78, "화학": 82}

성적표1.update(성적표2)
print(성적표1)
```

6. `copy()`: 딕셔너리를 복사하는 메서드

```python
bar = {"a": 100, "b": 200}
print(bar)

# 모든 원소 삭제
bar.clear()
print(bar)
```

## 연습문제

어느 학급의 수학점수가 다음과 같았다. 다음의 항목을 조건문/반복문을 사용하지 않고 구하시오.

"재석": 82
"민영": 91
"지수": 78
"영희": 85
"철수": 90

- 수학점수의 총 합은 얼마인가?
- 수학점수의 평균은 얼마인가?
- 수학점수의 최대값은 얼마인가?
- 수학점수의 최소값은 얼마인가?
- 수학점수의 범위(최대-최소)는 얼마인가?

```python
math = {"재석": 82, "민영": 91, "지수": 78, "영희": 85, "철수": 90}

s = list(math.values())

print("수학점수의 총 합:", sum(s))
print("수학점수의 평균:", sum(s) / len(s))
print("수학점수의 최대값:", max(s))
print("수학점수의 최소값:", min(s))
print("수학점수의 범위:", max(s) - min(s))
```
