### #02 csv 파일 읽기

1. csv 파일에서 썸네일 이미지 URL만 추출

```python
thumbnail_list = []

# CSV 파일을 읽어들인다
with open("카카오 책 검색(파이썬).csv", "r", encoding="utf-8") as f:
    # 행 단위로 분리해 리스트 형태로 읽어들임
    lines = f.readlines()

    for i, v in enumerate(lines):               # 행 단위 반복
        if i == 0: continue                     # 제목행 제외

        items = v.strip().split(",")            # 콤마로 분리

        for k in items:                         # 각 항목 반복
            if k.find("http") != -1:
                thumbnail_list.append(k)        # 썸네일 URL 리스트에 추가
                break                           # 다음 행으로 이동

# 썸네일 URL 리스트의 처음 5개를 출력
print(thumbnail_list[:5])
```

### #03 썸네일 이미지 다운로드

1. 다운로드 받은 이미지가 저장될 폴더 생성

```python
nowtime = dt.datetime.now()

dirname = nowtime.strftime("%Y%m%d_%H%M%S")

if not os.path.exists(dirname):
    os.makedirs(dirname)
    print(f"{dirname} 폴더가 생성되었습니다.")
```

```python
with requests.Session() as session:
    session.headers.update({
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36 Edg/147.0.0.0",
    })

    # 썸네일 이미지 리스트만큼 반복
    for i, v in enumerate(thumbnail_list):
        # 저장될 파일 경로
        fpath = "%s/%05d.jpg" % (dirname, i)

        try:
            # 썸네일 URL로 GET 요청을 보냄
            r = session.get(v, stream=True)

            if r.status_code != 200:
                msg = "%d번째 이미지 다운로드 에러 (%s)" % (i + 1, r.reason)
                # 다음 이미지 다운로드를 위해 continue로 넘어감
                continue

            # 다운로드 받은 파일 저장하기
            with open(fpath, "wb") as f:
                f.write(r.raw.read())
                print(f"[OK] {fpath} 저장 완료")

        except Exception as e:
            print(f"[ERROR] {fpath} 처리 실패: {e}")
```

## 정답

### 1. 데이터 수집하기

```python
import json
from pandas import DataFrame
```

```python
# 요청 주소
url = "https://data.hossam.kr/py/paiks_coffee_%03d.json"

# 전체 데이터 수
data_count = 19

# 수집 결과를 저장할 리스트
paiks = []

with requests.Session() as session:
    session.headers.update({
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    })

    # 전체 데이터 수 만큼 반복
    for i in range(1, data_count + 1):
        try:
            # 요청 주소에 페이지 번호를 넣어서 GET 요청을 보냄
            r = session.get(url % i)

            if r.status_code != 200:
                msg = "%d번째 데이터 다운로드 에러 (%s)" % (i, r.reason)
                print(msg)
                # 다음 데이터 다운로드를 위해 continue로 넘어감
                continue

            # JSON 형식으로 응답을 받음
            data = json.loads(r.text)

            # 수집 결과 리스트에 추가
            paiks.extend(data['stores'])

        except Exception as e:
            print(f"[ERROR] {i}번째 데이터 처리 실패: {e}")

df = DataFrame(paiks)
df.to_csv("paiks_coffee.csv", index=False, encoding="utf-8")

df
```

### 2. 매장 수 집계

```python
# 전체 매장 수
total_stores = len(paiks)

# 서울특별시에 위치한 매장 수
seoul_stores = 0

for store in paiks:
    if store['store_depth1'] == '서울특별시':
        seoul_stores += 1

# 서울특별시에 위치한 매장 수의 비율
seoul_ratio = (seoul_stores / total_stores) * 100

# 결과 출력
print(f"전체 매장 수: {total_stores}")
print(f"서울특별시에 위치한 매장 수: {seoul_stores}")
print(f"서울특별시에 위치한 매장 수의 비율: {seoul_ratio:.2f}%")
```

### 3. 이미지를 저장할 폴더 생성

```python
# 현재 날짜와 시간을 가져옴
nowtime = dt.datetime.now()

# 현재 날짜를 "YYYY-MM-DD" 형식으로 문자열로 저장
dirname = nowtime.strftime("%Y%m%d_%H%M%S")

# dirname 폴더 생성
if not os.path.exists(dirname):
    os.makedirs(dirname)
    print(f"{dirname} 폴더가 생성되었습니다.")
```

### 4. 서울특별시에 위치한 매장의 사진 내려받기

```python
with requests.Session() as session:
    session.headers.update({
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    })

    for store in paiks:
        if store['store_depth1'] == '서울특별시':
            # store_photo 다운로드
            photo_url = store['store_photo']
            fpath = "%s/%s.jpg" % (dirname, store['store_name'])

            try:
                r = session.get(photo_url, stream=True)

                if r.status_code != 200:
                    msg = "매장 ID %s 이미지 다운로드 에러 (%s)" % (store['store_id'], r.reason)
                    print(msg)
                    continue

                with open(fpath, "wb") as f:
                    f.write(r.raw.read())
                    print(f"[OK] {fpath} 저장 완료")
            except Exception as e:
                print(f"[ERROR] 매장 ID {store['store_id']} 이미지 처리 실패: {e}")
```
