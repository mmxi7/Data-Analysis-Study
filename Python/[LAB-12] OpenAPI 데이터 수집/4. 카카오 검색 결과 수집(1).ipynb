### #02 요청 보내기

1. 요청에 필요한 정보 준비하기

```python
# 요청 URL
url = "https://dapi.kakao.com/v3/search/book"

# QueryString 요청 변수
query = "파이썬"
page = 1
size = 50

# header에 포함할 키 입력
key = "dbeb4b2898f90cd053dd6b4d12cde8cc"
```

2. 웹 데이터 요청하기

```python
# 요청 결과를 저장하기 위한 리스트
resultset = []

with requests.Session() as session:
    session.headers.update({
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36 Edg/147.0.0.0",
        "Authorization": "KakaoAK %s" % key
    })

    for i in range(1, 51):
        r = session.get(url, params={"query": query, "page": i, "size": size})

        if r.status_code != 200:
            msg = "[%d Error] %s 에러가 발생함" % (r.status_code, r.reason)
        
            # 중간에 에러가 발생하더라도 중단하지 않고 다음 페이지 요청을 위해 계속 진행한다
            # raise Exception(msg)
            print(msg)
            continue

        # print(r)
        print("%d page 데이터 수신 완료 >> %s" % (i, r))

        r.encoding = "utf-8"
        mydict = r.json()
        resultset.extend(mydict["documents"])

print("총 %d건의 데이터가 수집됨" % len(resultset))
```
