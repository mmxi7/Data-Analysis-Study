2. 실습을 위한 샘플 데이터 준비

```python
point = np.array([100, 91, 89, 86, 79, 78, 77, 74, 71, 69, 66, 65, 60, 58, 57, 55])
point
```

### #02 NumPy의 histogram: 도수 분포도

1. 5개의 구간으로 나누기

```python
hist, bins = np.histogram(point, 5)
print("구간별 데이터 수:", hist)
print("구간 경계:", bins)
```

2. 도수 분포표 구성

```python
s = len(hist)
for i in range(0, s):
    if i + 1 < s:
        tpl = "%d이상 %d미만: %d개"
    else:
        tpl = "%d이상 %d미만: %d개"

    print(tpl % (bins[i], bins[i+1], hist[i]))
```

3. 데이터 구간을 직접 정의하기

```python
hist, bins = np.histogram(point, [50, 60, 70, 80, 90, 100])
print("구간별 데이터 수:", hist)
print("구간 경계:", bins)
```

### #03 도수의 종류

1. 절대 도수

2. 상대 도수

```python
relative_freq = hist / np.sum(hist)
relative_freq
```

3. 누적 도수

```python
cum_freq = []

previous = 0

for i in hist:
    previous += i
    cum_freq.append(previous)

cum_freq
```
