## 7. Heatmap

### #01 준비작업

1. 라이브러리 참조

```python
from jussam import load_data
from helpers import *
from pandas import pivot_table
import seaborn as sb
```

2. 데이터 가져오기

```python
origin = load_data('flights')
origin
```

3. 데이터 전처리(1) - 영문 월 이름을 숫자로 변환

```python
df = origin.copy()
df['month'] = df['month'].map({
    "January": 1, "February": 2, "March": 3, "April": 4,
    "May": 5, "June": 6, "July": 7, "August": 8,
    "September": 9, "October": 10, "November": 11, "December": 12
})

df.head()
```

4. 데이터 전처리(2) - 히트맴에 적합한 형태로 데이터 재구조화

```python
df2 = pivot_table(df, index="year", columns="month",
                  values="passengers")
df2
```

5. 데이터 전처리(3) - 년도 역순으로 정렬

```python
df3 = df2.sort_index(ascending=False)
df3
```

### #02 Heatmap 시각화

1. 년도/월별 탑승객 수 분포 시각화

```python
fig, ax = my_plot.init(width=1200, height=600,
                       title="Heatmap of Passengers by Year and Month",
                       xlabel="Month", ylabel="Year")

ax.grid(False)  # 격자 제거

# 2) Heatmap 그리기
# -> annot=True : 수치값을 함께 표시함
# -> fmt: annot=True가 설정된 경우 표시되는 수치값의 형식 지정
# -> linewidth: 각 셀 사이의 선 굵기
# -> cmap: 칼라맵 ('Greys' 등)
sb.heatmap(data=df3, annot=True, fmt="0.1f", linewidths=0.5, cmap="GnBu")

my_plot.show()
```

> ### 인사이트
> - 연도 증가에 따라 전반적 승객 수가 뚜렷하게 증가함.
>   - 예) 1949년 7월 148 → 1960년 7월 622
> - 계절 패턴이 반복적으로 나타남:
>   - 매년 7\~8월이 고점,
>   - 11\~2월이 저점 구간임.
> - 전체 최댓값은 1960년 7월 622로 확인됨.
> - 전체 최솟값은 1949년 11월 104로 확인됨.
> - 같은 월 기준으로도 연도별 상승 흐름이 일관됨.
>   - 예) 1월 112→417, 12월 118→432
> - 고연도로 갈수록 고점과 저점의 차이가 커져, 계절 변동 폭도 함께 확대되는 모습이 보임.

### 교통사고 현황 데이터의 인사이트 찾기

```python
# 라이브러리 참조
from jussam import load_data
from helpers import *
from pandas import pivot_table
import seaborn as sb
```

```python
# 데이터 불러오기
origin = load_data('traffic_acc')
origin
```

```python
# 데이터 재구조화
df = pivot_table(origin, index="년도", columns="월",
                  values="발생건수")

# 년도 역순으로 정렬
df1 = df.sort_index(ascending=False)
df1
```

```python
# 히트맵 그리기
my_plot.heatmap(df1, width=1400, height=600,
                title="년도에 따른 월별 교통사고 발생건수",
                xlabel="월", ylabel="년도")
```

2. 히트맵 전체 기간 중에서 색이 가장 진하게 나타난, 최대 발생건수를 기록한 년-월은 2015년 10월이다.
