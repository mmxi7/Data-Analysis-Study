## 문제 10. 나이(`age`)가 많아질수록 임금(`wage`)은 대체로 어떻게 변하는가?

① 계속 가파르게 증가한다  ② 중년까지 증가하다 이후 정체·감소하는 경향  ③ 계속 감소한다  ④ 아무런 경향이 없다

## 문제 11. 연령대별 평균 임금이 가장 높은 구간은?

① 20대 이하  ② 30대  ③ 40대  ④ 60대 이상

> 💡 **힌트**: 산점도(lmplot) 또는 연령대 구간화 후 막대그래프
> 🎯 **ADsP 개념**: 연속형 변수의 구간화(binning) · 비선형 추세

```python
origin['age_bins'] = pd.cut(origin['age'], bins=[0, 20, 30, 40, 50, 60, 81], right=False, 
                  labels=['20대 미만', '20대', '30대', '40대', '50대', '60대 이상'])

my_plot.barplot(data=origin, x="age_bins", y="wage",
                estimator=np.mean, palette="Set2",
                xlabel="연령대", ylabel="평균 임금")
```

## 문제 14. 고임금자(`wage` 250 초과)는 어느 직군에 더 많이 분포하는가?

① 생산직(Industrial)  ② 사무직(Information)  ③ 동일하다  ④ 생산직에만 있다

> 💡 **힌트**: 막대그래프(countplot, wage>250 필터 후 jobclass) 또는 누적막대그래프
> 🎯 **ADsP 개념**: 조건부 필터링 · 교차 빈도 분석

```python
origin['wage_250'] = np.where(origin['wage'] > 250, '250 초과', '250 이하')
```

```python
my_plot.stackplot(data=origin,                  # 데이터 프레임
                  x="jobclass",                      # x축에 사용할 열 이름
                  y="wage",                  # y축에 사용할 열 이름
                  hue="wage_250",                    # 누적 막대그래프의 범주형 열 이름
                  aggfunc=len,               # 집계 함수 (기본값: np.sum)
                  orient='h',                   # 그래프 방향 ('v' -> 수직, 'h' -> 수평)
                  text_format="{:d}명",         # 텍스트 포맷 (기본값: "{:.1f}")
                  palette="tab20",              # 색상 팔레트 (기본값: None)
                  title="직군별 임금",            # 그래프 제목 (기본값: None)
                  xlabel="총 종사자 수",          # x축 레이블 (기본값: None)
                  ylabel="직군")                 # y축 레이블 (기본값: None)
```

- ② 사무직(Information)
