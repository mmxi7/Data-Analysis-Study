### #03 기술 통계량

1. ndarray의 기능을 그대로 지원

```python
column = Series([188, 92, 45, 63, 79])

print("합계:", column.sum())
print("평균:", column.mean())
print("최대:", column.max())
print("최소:", column.min())
print("분산:", column.var())
print("표준편차:", column.std())
print("중앙값:", column.median())
print("1사분위 수:", column.quantile(0.25))
print("2사분위 수:", column.quantile(0.5))
print("3사분위 수:", column.quantile(0.75))
```

### #04 조건 검색

1. 샘플 데이터 정의

```python
column = Series(np.arange(0, 100, 10))
column
```

2. NumPy와 동일한 표현식 사용

```python
# numpy와 같은 표현식을 사용한다.
in1 = column[column > 30]
in1
```

3. AND 검색

```python
in2 = column[column <= 70][column > 10]
in2
```

4. OR 검색

```python
in3 = column[(column <= 10) | (column >= 70)]
in3
```

### #06 데이터 정렬

1. 값 기반 정렬(오름차순)

```python
my_dict = {"c": 456, "a": 123, "b": 789}
origin = Series(my_dict)

sort1 = origin.sort_values()
sort1
```

2. 값 기반 정렬(내림차순)

```python
sort2 = origin.sort_values(ascending=False)
sort2
```

3. 인덱스 기반 정렬(오름차순)

```python
sort3 = origin.sort_index()
sort3
```

4. 인덱스 기반 정렬(내림차순)

```python
sort4 = origin.sort_index(ascending=False)
sort4
```

## 연습문제

다음은 어떤 학생의 5과목에 대한 시험 점수를 딕셔너리로 나타낸 것입니다.   
   
scores = {"국어": 85, "영어": 92, "수학": 78, "과학": 95, "사회": 88}   
   
Pandas의 Series를 사용하여 다음 작업을 수행하는 코드를 작성하세요.   

1. 위 딕셔너리를 Series 객체로 변환하세요.   
2. 수학 과목의 점수를 출력하세요.   
3. 점수가 90점 이상인 과목만 선택하여 출력하세요.   
4. 모든 과목 점수의 평균을 계산하여 출력하세요.

```python
import numpy as np
from pandas import Series

scores = {"국어": 85, "영어": 92, "수학": 78, "과학": 95, "사회": 88}

score_series = Series(scores)
print("--- 전체 점수 ---")
print(score_series)

print("--- 수학 점수 ---")
print(f"수학 점수: {score_series["수학"]}")

print("--- 90점 이상 과목 ---")
print(score_series[score_series >= 90])

print("--- 평균 점수 ---")
print(f"평균 점수: {score_series.mean()}")
```
