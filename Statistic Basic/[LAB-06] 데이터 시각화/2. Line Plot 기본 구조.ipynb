- 딸기는 07-01부터 07-03까지 뚜렷한 상승 뒤 07-04에 하락하는 패턴임
- 초콜릿과 딸기 비교 시, 07-02 이후에는 딸기 판매량이 초콜릿을 상회함
- 민트초코는 표시된 구간(07-02 ~ 07-04)에서 매 시점 꾸준히 증가함
- 데이터 표시 기준으로 일부 맛은 모든 날짜에 값이 있지 않음: 바닐라는 3개 시점, 민트초코도 3개 시점만 시각화되어 있음

## 2-2. Line Plot 옵션 설정

### #01 준비작업

1. 라이브러리 참조 (략)

2. 데이터셋 불러오기 (략)

### #02 선 모양 설정하기

1. 선 색상 | 선 스타일, 굵기 | 마커 모양, 크기, 테두리 색상, 면 색상, 테두리 굵기

```python
# 그래프 초기화
fig, ax = my_plot.init(title="아이스크림 판매량", xlabel="날짜", ylabel="판매수량")

# LinePlot 그리기
sb.lineplot(data=origin, x='Date', y='Quantity',
            c="#1D84C3", linestyle=':', linewidth=5,
            marker="o", markersize=20, markeredgewidth=3,
            markerfacecolor="#00ff00", markeredgecolor="#ff0000")

# 출력
my_plot.show()
```

### #03 축 설정

1. 축 범위 / 표시할 값 설정하기

```python
# 그래프 초기화
fig, ax = my_plot.init(title="아이스크림 판매량", xlabel="날짜", ylabel="판매수량")

# LinePlot 그리기
sb.lineplot(data=origin, x='Date', y='Quantity')

# y축 범위 (x축 범위 설정은 set_xlim)
ax.set_ylim(18, 28)

# y축에 표시할 값 설정
# --> [18~28]에 표시할 내용을 [18~28] 범위의 실제 숫자로 지정
# --> 글자크기=6, 표시각도=-45도
ax.set_yticks(list(range(18, 29)), list(range(18, 29)), fontsize=6, rotation=-45)

# 출력
my_plot.show()
```

### #04 다중 선 그래프

1. 두 개 이상의 선 그리기

```python
# 그래프 초기화
fig, ax = my_plot.init(title="범주별 그래프 그리기", xlabel="X", ylabel="Y")

# 그래프 그리기
sb.lineplot(x=[1, 2, 3, 4], y=[10, 20, 35, 40], label="A변화량")
sb.lineplot(x=[1, 2, 3, 4], y=[10, 15, 30, 35], label="B변화량")
ax.legend(loc='upper left')         # 범주의 위치 설정 (미설정 시 자동으로 위치가 지정됨)

# 출력
my_plot.show()
```

- 토핑별 시계열 길이가 다름: 초코시럽 2개 시점, 초코칩은 3개 시점만 표시되어 일부 날짜가 비어있음

### Covid19 확진자 수 변동 추이 시각화

```python
# 라이브러리 참조하기
from jussam import load_data
from helpers import my_plot
from pandas import melt
import seaborn as sb
```

```python
# 데이터 불러오기
active = load_data('covid19_active')
active
```

```python
# hue를 사용하기 위한 데이터 재구조화
pivot_active = melt(active.reset_index(), id_vars=['기준일'], value_vars=['서울시 일일 확진', '전국 일일 확진'], var_name='지역', value_name='확진자 수')
pivot_active
```

```python
my_plot.lineplot(data=pivot_active, x='기준일', y="확진자 수", hue="지역",
                 title="지역 별 확진자 수", xlabel="기준일", ylabel="확진자 수")
```

- 조사 기간 중 전국 일일 확진자 수가 가장 많았던 시점은 2022년 9월이다.
- 그래프에서 확진자 수의 큰 봉우리는 총 2번 나타났다.
- 서울시와 전국의 확진자 수 중, 항상 더 많은 쪽은 전국의 확진자 수이다.
