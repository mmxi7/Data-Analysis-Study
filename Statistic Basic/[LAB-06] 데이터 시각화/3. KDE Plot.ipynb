### #02 일변량 커널 밀도

1. 단일 분포 -> 하나의 변수에 대한 KDE Plot

```python
# 그래프 초기화
fig, ax = my_plot.init()

# 그래프 그리기 (아래 패턴 3가지는 모두 결과가 동일함)
# [패턴1] 자동인식 패턴
# sb.kdeplot(data=origin["Sepal.Width"])
# [패턴2] 개별 데이터 지정 패턴
# sb.kdeplot(x=origin["Sepal.Width"])
# [패턴3] DataFrame 지정 패턴
sb.kdeplot(data=origin, x="Sepal.Width", linewidth=2)

# 출력
my_plot.show()
```

> 인사이트
- Sepal.Width는 약 1.5 ~ 4.9 범위에 분포한다.
- 분포는 단봉(unimodal)이며, 가장 밀도가 높은 값은 약 3.0이다.
- 값의 대부분은 2.5 ~ 3.5 구간에 집중되어 있다.
- 양쪽 극단(2.0 이하, 4.0 이상)에서는 관측 밀도가 매우 낮다.

2. 다중 분포

```python
# 그래프 초기화
fig, ax = my_plot.init()

# 그래프 그리기 - x 대신 y를 설정할 경우 가로로 표시된다.
sb.kdeplot(data=origin)

# 출력
my_plot.show()
```

3. 옵션 설정

```python
# 그래프 초기화
fig, ax = my_plot.init()

# 그래프 그리기
sb.kdeplot(data=origin,         # 데이터 연결
           x="Sepal.Length",     # 분포를 확인할 컬럼 지정
           hue="Species",       # 범주 지정
           fill=True,           # 면 색상 채움
           alpha=0.3,           # 면 색상의 투명도
           palette="pastel",    # 색상 팔레트
           linewidth=2          # 선의 두께
)

# 출력
my_plot.show()
```

> 인사이트
- setosa의 Sepal.Length는 가장 짧은 구간에 분포하며, 약 5.0 부근에서 단일 피크를 가진다.
- versicolor는 setosa보다 큰 값 구간에 분포하며, 약 5.8 ~ 6.0 부근에서 피크가 나타난다.
- virginica는 세 종 중 가장 큰 Sepal.Length 값을 가지며, 약 6.5 부근에서 피크를 보인다.
- 세 종의 분포는 중심 위치가 순차적으로 이동한다. (setosa > versicolor > virginica)
- setosa와 다른 두 종 간의 분포 겹침은 제한적이다.
- versicolor와 virginica는 일부 구간(약 6.0 전후)에서 분포가 겹친다.

### #03 이변량 커널 밀도

1. x에 따른 y 분포

```python
# 그래프 초기화
fig, ax = my_plot.init()

# 그래프 그리기
sb.kdeplot(data=origin, x="Petal.Width", y="Sepal.Width", hue="Species")

# 출력
my_plot.show()
```

> 인사이트
- Petal.Width 기준으로 setosa는 완전히 분리되며, versicolor와 virginica는 부분적으로 겹침
- setosa(파란색):   
    - 가장 좁은 꽃잎 폭 (0 ~ 0.6) & 중간 정도의 꽃받침 폭(2.0 ~ 4.6)   
        - 가장 왼쪽에 집중된 분포로 다른 종과 명확하게 분리됨   
- versicolor(주황색):   
    - 중간 크기의 꽃잎 폭 (0.8 ~ 1.9) & 가장 좁은 꽃받침 폭 (1.8 ~ 3.6)   
        - setosa와 virginica 사이의 중간 특성을 보유   
        - 두 종 사이에 겹치는 영역이 존재
- virginica(초록색):   
    - 가장 넓은 꽃잎 폭 (1.3 ~ 2.7) & 중간 꽃받침 폭 (2.0 ~ 4.0)   
        - 오른쪽 상단에 위치하여 다른 종과 구분됨

### #04 커널밀도와 평균값을 함께 표시하기

1. 평균값 구하기

```python
mean_sepal_width = origin["Sepal.Width"].mean()
print(mean_sepal_width)
```

2. 커널밀도 + 평균값

```python
# 그래프 초기화
fig, ax = my_plot.init()

# 그래프 그리기
sb.kdeplot(data=origin, x="Sepal.Width")

# 평균에 대한 수직선 추가
ax.axvline(x=mean_sepal_width, linestyle='--', color='#ff2200')

# 텍스트 표시하기
ax.text(x=mean_sepal_width + 0.05, y=ax.get_ylim()[1] * 0.95,
        s=f'Mean: {mean_sepal_width:.2f}', color='#ff2200', fontsize=14)

# 출력
my_plot.show()
```

> 인사이트
- Sepal.Width의 분포는 대칭적인 종 모양의 단봉형 분포를 나타냄
- 평균값(Mean: 3.06)이 분포의 최고점과 일치함
    - 중앙에 위치한 수직선이 커널밀도 곡선의 피크와 겹침
    - 데이터가 평균을 중심으로 균령 잡혀 분포함을 의미
- 분포의 범위는 약 1.5 ~ 5.0 사이
    - 2.5 ~ 3.5 구간에 가장 높은 밀도로 관찰됨
    - 양쪽 끝으로 갈 수록 밀도가 점진적으로 감소
- 정규분포에 가까운 형태
    - 대칭성이 뚜렷하고 이상치가 거의 없음을 시사
