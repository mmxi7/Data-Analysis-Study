### #04 인덱스를 활용한 병합

1. 학생의 이름을 인덱스로 갖는 두 데이터 프레임 생성

```python
df_left = DataFrame({'수학': [90, 82]}, index=['민철', '봉구'])
df_left
```

```python
df_right = DataFrame({'국어': [90, 82]}, index=['민철', '철수'])
df_right
```

2. 병합 조건에서 index를 사용하도록 지정하여 Inner Join하기

```python
merge(df_left, df_right, left_index=True, right_index=True)
```

3. 병합 조건에서 index를 사용하도록 지정하여 Left Outer Join하기

```python
merge(df_left, df_right, left_index=True, right_index=True, how="left")
```

4. 병합 조건에서 index를 사용하도록 지정하여 Right Outer Join하기

```python
merge(df_left, df_right, left_index=True, right_index=True, how="right")
```

5. 병합 조건에서 index를 사용하도록 지정하여 Full Outer Join하기

```python
merge(df_left, df_right, left_index=True, right_index=True, how="outer")
```

### #05 인덱스와 컬럼을 각각 기준으로 하기

1. 샘플 데이터 프레임 생성

```python
df_left = DataFrame({'수학': [90, 82]}, index=['민철', '봉구'])
df_left
```

```python
df_right = DataFrame({'성명': ['민철', '철수'], '영어': [90, 82]})
df_right
```

2. 왼쪽은 인덱스, 오른쪽은 컬럼을 기준으로 Inner Join

```python
merge(df_left, df_right, left_index=True, right_on=['성명'])
```

3. 왼쪽은 인덱스, 오른쪽은 컬럼을 기준으로 Left Outer Join

```python
merge(df_left, df_right, left_index=True, right_on=['성명'], how='left')
```

4. 왼쪽은 인덱스, 오른쪽은 컬럼을 기준으로 Right Outer Join

```python
merge(df_left, df_right, left_index=True, right_on=['성명'], how='right')
```

5. 왼쪽은 인덱스, 오른쪽은 컬럼을 기준으로 Full Outer Join

```python
tmp = merge(df_left, df_right, left_index=True, right_on=['성명'], how='outer')
tmp
```

6. 병합 후에는 인덱스를 정리해야 한다.

```python
tmp2 = tmp.set_index('성명')
tmp2
```

### 1. 고객 데이터와 구매 데이터를 통합하여, 연령대별 평균 구매액 분석하기

- 팀장님께서 고객 정보와 최근 구매 내역 데이터를 전달해 주셨습니다. 하지만 데이터가 완벽하지는 않네요. 여러분의 데이터 분석 스킬을 발휘하여 아래 요구사항에 따라 데이터를 정제하고 분석 리포트를 만들어주세요.

- 고객 정보 > 직접 생성해야 합니다.

| 변수명 | 데이터 |
|---|---|
| `customer_id` | 101, 102, 103, 104, 105, 106 |
| `name` | 'Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank' |
| `age` | 25, 32, 45, 38, 29, 41 |
- 구매내역: `order_history` 데이터셋으로 가져올 수 있습니다.

- 요구사항

1. 구매 데이터의 `price` 컬럼에 있는 결측치는 전체 상품의 **평균 가격**으로 채워주세요.

2. 정제된 두 데이터프레임을 `customer_id` 기준으로 병합하세요.

3. 최종 병합된 데이터 프레임에 대한 **기술 통계량**을 도출하세요.

```python
# 1. 라이브러리 참조
from jussam import load_data
from pandas import DataFrame
from pandas import merge, concat

# 2. 구매 내역 데이터 불러오기
order_df = load_data('order_history')

# 3. 고객 정보 생성
data = {
    "customer_id": [101, 102, 103, 104, 105, 106],
    "name": ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    "age": [25, 32, 45, 38, 29, 41]
}

customer_df = DataFrame(data)

display(order_df)
display(customer_df)
```

```python
# 4. 구매 데이터 결측치 대체
#imr = SimpleImputer(missing_values=np.nan, strategy='mean')
#reshape = order_df['price'].values.reshape(-1, 1)
#reshape_imr = imr.fit_transform(reshape)
#order_df['price'] = reshape_imr

결측치처리 = order_df.copy()        # 원본 복사
average_price = 결측치처리['price'].mean()      # 전체 상품의 평균 가격
결측치처리_결과 = 결측치처리['price'].fillna(average_price)     # 평균 가격으로 결측치 대체
결측치처리['price'] = 결측치처리_결과       # Series 객체를 원본 데이터프레임의 price 컬럼에 대입

결측치처리
```

```python
# 5. 두 데이터 프레임을 customer_id 기준으로 병합
final_df = merge(결측치처리, customer_df, on='customer_id')
final_df
```

```python
# 6. 최종 데이터 프레임에 대한 기술 통계량 도출
final_df.describe().T.drop('customer_id')
```

### 2. 흩어진 판매 데이터 통합하기

당신은 IT 쇼핑몰의 신입 데이터 분석가입니다.

- 회사에는 강남점과 종로점, 두 개의 지점이 있습니다.
- 각 지점의 판매 데이터가 별도의 데이터(`shop_강남`, `shop_종로`)로 관리되고 있었습니다.

연말 보고를 위해 두 지점의 데이터를 하나로 합쳐 **전체 판매 실적을 분석**해야 합니다.

- 하지만 데이터를 살펴보니 몇 가지 정리할 부분이 보입니다.
- 당신의 임무는 데이터를 정제하고 통합하여 의미있는 리포트를 만드는 것입니다.

#### 요구사항

1. 강남점 데이터에는 `Mouse` 상품의 가격(`Price`)이 누락되어 있습니다.  
   이 값을 `8000`으로 채워주세요.

2. 종로점 데이터에는 `Keyboard` 상품의 주문 수량(`Quantity`)이 `999`로 잘못 입력되어 있습니다.  
   실제 수량은 `9`이므로, 이 값을 수정해주세요.

3. 두 개의 `DataFrame`을 하나의 `DataFrame`으로 합치세요.  
   합치는 과정에서 기존의 인덱스는 의미가 없으므로, 새로운 인덱스를 부여하세요.

4. 화면에 최종 `DataFrame`의 전체 내용을 출력하세요.

```python
# 1. 라이브러리 참조
import numpy as np
from jussam import load_data
from pandas import DataFrame
from sklearn.impute import SimpleImputer

# 2. 데이터 불러오기
shop_gn = load_data('shop_강남')
shop_jl = load_data('shop_종로')

display(shop_gn)
display(shop_jl)
```

```python
# 3. 강남점 결측치를 8000으로 대체
re_shop_gn = shop_gn.copy()
re_shop_gn.loc[1, 'Price'] = 8000
re_shop_gn
```

```python
# 4. 종로점 데이터 수정하기
shop_jl_cp = shop_jl.copy()
shop_jl_cp.loc[shop_jl.Product == "Keyboard", "Quantity"] = 9
shop_jl_cp
```

```python
# 5. 두 개의 데이터 프레임을 병합하고 새로운 인덱스 부여
final_df = concat([re_shop_gn, shop_jl_cp], ignore_index=True)
```
