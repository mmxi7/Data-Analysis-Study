### #01 준비작업

1. 라이브러리 참조

```python
from jussam import load_data
```

2. 데이터 불러오기

```python
origin = load_data('grade')
origin
```

### #02 데이터 정렬

1. 변수에 의한 정렬

```python
단일정렬df = origin.sort_values('학년', ascending=False)
단일정렬df
```

2. 여러 변수 정렬

```python
복수정렬df = origin.sort_values(['학년', '국어'], ascending=[False, False])
복수정렬df
```

3. 인덱스를 기준으로 하는 정렬

```python
인덱스정렬df = origin.sort_index(ascending=False)
인덱스정렬df
```

### #03 데이터 필터링

1. 숫자형 변수의 조건 검색

```python
# r1 = origin[origin['국어'] > 80]
r1 = origin.query('국어 > 80')
r1
```

2. 문자열 변수의 조건 검색

```python
# r2 = origin[origin["성별"] == '남자]
r2 = origin.query("성별 == '남자'")
r2
```

3. AND 연산을 사용하는 조건 검색

```python
# r3 = origin[(origin['국어'] > 80) & (origin['수학']) > 80]
r3 = origin.query('국어 > 80 and 수학 > 80')
r3
```

4. OR 연산을 사용하는 조건 검색

```python
# r4 = origin[(origin['국어'] <= 80) | (origin['수학']) <= 80]
r4 = origin.query('국어 <= 80 or 수학 <= 80')
r4
```

5. 비교식에 변수를 사용하는 조건 검색

```python
point = int(input('점수를 입력하세요: '))
print("국어 점수가 %d점보다 높은 학생" % point)

r5 = origin.query('국어 > @point')          # @변수이름 형태로 조건식에 포함
r5
```

6. 열 단위 필터링

```python
fl = origin.filter(['국어', '영어'])
fl
```

### 마케팅팀 신입사원의 첫 번째 미션

**VIP 고객을 찾아라!**
- 당신은 방금 우리 회사 마케팅팀에 입사한 신입사원입니다.
- 팀장님께서 당신에게 첫 번째 미션을 주셨습니다. 연말을 맞이하여 VIP 고객들에게 감사의 마음을 담은 특별 선물을 보내려고 합니다.
- 당신의 임무는 제공된 고객 데이터 vip_customer를 분석하여 VIP 고객 후보 명단을 만드는 것입니다.   

**아래의 조건들을 모두 만족하는 진정한 VIP 고객을 찾아내어, 그들의 명단을 팀장님께 보고해야 합니다.**
1. 우수 고객: 총 구매액(TotalSpending)이 100,000원 이상인 고객
2. 활동지역 고객: 서울(City)에 거주하는 고객   

**보고할 내용**
- 위 세 가지 조건을 모두 만족하는 고객들을 총 구매액(TotalSpending)이 높은 순서대로 정렬하여 상위 5명의 고객 정보를 추출하세요.
- 추출된 5명의 고객 데이터에서 Name, City, TotalSpending만 선택하여 최종 명단을 만드세요.

```python
from jussam import load_data

vip = load_data('vip_customer')

vip_filtered = vip.query('TotalSpending >= 100000 and City == "서울"')
vip_sorted = vip_filtered.sort_values('TotalSpending', ascending=False).head()
vip_final = vip_sorted.filter(['Name', 'City', 'TotalSpending'])

vip_final
```
