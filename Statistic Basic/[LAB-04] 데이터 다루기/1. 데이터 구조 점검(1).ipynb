```python
컬럼단위_명목df = origin.copy()
컬럼단위_명목df['학년'] = 컬럼단위_명목df['학년'].astype('category')
컬럼단위_명목df['성별'] = 컬럼단위_명목df['성별'].astype('category')
컬럼단위_명목df.info()
```

6. 범주형 데이터를 명목척도로 변환(2)

```python
범주형df = origin.astype({'학년': 'category', '성별': 'category'})
범주형df.info()
```

7. 범주형 데이터를 서열척도로 변환(1)

```python
컬럼단위_서열척도df = origin.copy()
컬럼단위_서열척도df['학년'] = 컬럼단위_서열척도df['학년'].astype(
    CategoricalDtype(categories=[1, 2, 3, 4],       # 범주형 값의 종류를 담은 리스트
    ordered=True)                                   # 서열척도임을 표시, 기본값은 False(명목척도)
    )
컬럼단위_서열척도df.info()
```

8. 범주형 데이터를 서열척도로 변환(2)

```python
서열척도df = origin.astype({
    "학년": CategoricalDtype(categories=[1, 2, 3, 4], ordered=True)
})
서열척도df.info()
```

### 2. 데이터 구성하기
네 종류의 식단 A1, A2, A3, A4에 따른 식이요법이 혈액응고시간에 영향을 주는지 알고자 실험한 자료가 다음과 같다.   
- 인자의 수준을 나타내는 변수를 class로, 실험값을 time으로 하여 데이터를 구성하여라.   
- class 변수는 명목형 변수이다.   
- 데이터를 직접 코드에 명시해야 합니다. 

| 인자의 수준 | 실험의 반복 |
|---|---|
| A1 | 62, 60, 63, 59, 61 |
| A2 | 62, 67, 71, 64, 65, 66 |
| A3 | 68, 66, 71, 67, 68, 69, 70 |
| A4 | 58, 62, 60, 59, 61, 63, 64 |

```python
from pandas import DataFrame, CategoricalDtype

data = {
    "class": ["A1", "A1", "A1", "A1", "A1", "A2", "A2", "A2", "A2", "A2", "A2", "A3", "A3", "A3", "A3", "A3", "A3", "A3", "A4", "A4", "A4", "A4", "A4", "A4", "A4"],
    "time": [62, 60, 63, 59, 61, 62, 67, 71, 64, 65, 66, 68, 66, 71, 67, 68, 69, 70, 58, 62, 60, 59, 61, 63, 64]
}

df = DataFrame(data)
typed_df = df.astype({'class': 'category'})

typed_df
```

```python
blood_data = [
    [62, 60, 63, 59, 61],
    [62, 67, 71, 64, 65, 66],
    [68, 66, 71, 67, 68, 69, 70],
    [58, 62, 60, 59, 61, 63, 64]
]

class_name = ['A1', 'A2', 'A3', 'A4']

class_list = []
time_list = []

for i in range(len(blood_data)):
    for value in blood_data[i]:
        class_list.append(class_name[i])
        time_list.append(value)

df = DataFrame({
    'class': class_list,
    'time': time_list
})

df
```
