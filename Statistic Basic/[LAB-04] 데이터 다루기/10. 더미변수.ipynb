## 10. 더미변수

### #01 준비작업

1. 라이브러리 참조

```python
from jussam import load_data

# 데이터프레임, 더미변수 생성 함수, 데이터 병합 함수
from pandas import DataFrame, get_dummies, merge

# 원-핫 인코딩 클래스 (머신러닝용 더미변수)
from sklearn.preprocessing import OneHotEncoder
```

2. 데이터 불러오기

```python
origin = load_data('nursing_grades')
origin
```

3. 범주형 타입 변환

```python
타입변환_df = origin.astype({'성별': 'category', '합격여부': 'category'})
타입변환_df.info()
```

### #02 데이터 라벨링

1. 성별에 대한 값 종류 확인

```python
vcounts = 타입변환_df['성별'].value_counts()
values = vcounts.sort_index()
values
```

2. 성별에 대한 라벨링용 딕셔너리 생성

```python
labels = {}

for i, v in enumerate(values.index):
    labels[v] = i

labels
```

3. 성별 변수에 라벨링 적용

```python
라벨링_df = 타입변환_df.copy()
라벨링_df['성별'] = 라벨링_df['성별'].map(labels)
라벨링_df['성별'].value_counts().sort_index()
```

4. '합격여부' 변수에 대한 일괄 처리

```python
vcounts = 라벨링_df['합격여부'].value_counts()
values = vcounts.sort_index()

for i, v in enumerate(values.index):
    labels[v] = i

라벨링_df['합격여부'] = 라벨링_df['합격여부'].map(labels)
라벨링_df['합격여부'].value_counts().sort_index()
```

5. 라벨링이 적용된 데이터 프레임 확인

```python
라벨링_df
```

### #03 Pandas를 사용하는 더미변수 생성 방법

1. 하나의 컬럼에 대해 모든 값을 더미변수로 변환 (값의 수에 따라 N개 생성)

```python
더미변수_df_1 = get_dummies(라벨링_df, columns=['성별'], dtype='int')
더미변수_df_1.head()
```

2. 두 개 이상의 명목형 변수 처리

```python
더미변수_df_2 = get_dummies(라벨링_df, columns=['성별', '합격여부'], dtype='int')
더미변수_df_2.head()
```

3. N-1 개의 더미변수 생성

```python
더미변수_df_3 = get_dummies(라벨링_df, columns=['성별', '합격여부'], dtype='int', drop_first=True)
더미변수_df_3.head()
```

### #04 Scikit-Learn을 사용한 단일 컬럼의 인코딩

1. 인코딩 대상 변수 추출

```python
X = 타입변환_df[['성별']]
X.head()
```

2. OneHotEncoding 처리

```python
encoder = OneHotEncoder(sparse_output=False, drop=None, dtype=int)
result = encoder.fit_transform(X)

# 출력량이 많으므로 10개만 출력
result[:10]
```

3. 인코딩 결과를 데이터 프레임으로 생성

```python
# 컬럼명 생성
new_cols = encoder.get_feature_names_out(['성별'])
new_cols
```

```python
# DataFrame 병합
one_hot_df = DataFrame(result, columns=new_cols, index=타입변환_df.index)
one_hot_df.head()
```

4. 원본 데이터 프레임과 병합

```python
인코딩_df_1 = merge(타입변환_df, one_hot_df, left_index=True, right_index=True)
인코딩_df_1.head()
```

5. 불필요한 컬럼 제거

```python
인코딩_df_1_최종 = 인코딩_df_1.drop(columns='성별')
인코딩_df_1_최종.head()
```

### #04 Scikit-Learn을 사용한 다중 컬럼의 인코딩

1. 여러 컬럼에 대한 일괄 처리

```python
X = 타입변환_df[['성별', '합격여부']]

encoder = OneHotEncoder(sparse_output=False, drop=None, dtype=int)
result = encoder.fit_transform(X)

result
```

2. 컬럼명 생성

```python
new_cols = encoder.get_feature_names_out(['성별', '합격여부'])
new_cols
```

3. 데이터 프레임으로 병합

```python
one_hot_df = DataFrame(result, columns=new_cols, index=타입변환_df.index)
one_hot_df.head()
```

4. 원본 데이터 프레임과 병합

```python
인코딩_df_2 = merge(타입변환_df, one_hot_df, left_index=True, right_index=True)
인코딩_df_2.head()
```

5. 불필요한 컬럼 제거

```python
인코딩_df_2.drop(columns=['성별', '합격여부'], inplace=True)
인코딩_df_2.head()
```

### 🚀 온라인 스토어 고객 행동 분석
**□ 한 온라인 스토어의 마케팅 팀이 연말 특별 프로모션을 기획하고 있습니다.**   

◎ 프로모션의 효과를 극대화하기 위해, 고객 데이터를 분석하여 타겟 고객층을 식별하고자 합니다.   
◎ 여러분은 데이터 분석가로서 마케팅 팀의 의사결정을 돕기 위해 고객 데이터를 전처리하고 분석용 데이터셋을 만드는 임무를 맡았습니다.   

**□ 주어지는 데이터 셋**   

◎ online_store_customers : 고객의 기본 인구 통계 정보   
◎ online_store_purchases : 고객의 구매 관련 정보   

1. 두 데이터를 불러와서 고객ID를 기준으로 병합하세요.
2. 구매 데이터의 'size' 열에 존재하는 결측치는 가장 빈도가 높은 사이즈로 채워주세요.
3. 성별에 따른 구매 패턴을 분석하기 위해 'gender' 열에 대해 get_dummies() 함수를 사용하여 더미변수를 생성하세요. (라벨링은 생략합니다.)
4. 색상 선호도를 분석하기 위해 'color' 열을 원-핫 인코딩(One-hot encoding) 방식으로 변환하여 새로운 열들을 추가하세요. (라벨링은 생략합니다.)

```python
# 1. 라이브러리 참조
from jussam import load_data
from pandas import DataFrame, get_dummies, merge
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
import numpy as np

# 2. 데이터 불러오기
customers = load_data('online_store_customers')
purchases = load_data('online_store_purchases')

display(customers, purchases)
```

```python
# 3. 고객ID 기준으로 병합
merged_df = merge(customers, purchases, left_index=True, right_on='user_id')
merged_df
```

```python
# 4. 데이터 정제 (size의 결측치를 최빈도로 대체)
cleaned_df = merged_df.copy()
most_frequent = merged_df['size'].mode().iloc[0]
cleaned = merged_df['size'].fillna(most_frequent)

cleaned_df['size'] = cleaned
cleaned_df
```

```python
# 5. 성별 컬럼에 더미 변수 생성
dm_df1 = get_dummies(cleaned_df, columns=['gender'], dtype='int')
dm_df1.head()
```

```python
# 6. 컬러 컬럼을 원-핫 인코딩으로 변환해 새로운 열 추가
X = dm_df1[['color']]

encoder = OneHotEncoder(sparse_output=False, drop="first", dtype=int)
result = encoder.fit_transform(X)

new_cols = encoder.get_feature_names_out(['color'])

one_hot_df = DataFrame(result, columns=new_cols, index=dm_df1.index)

인코딩_df = merge(dm_df1, one_hot_df, left_index=True, right_index=True)

인코딩_df_최종 = 인코딩_df.drop(columns='color')
인코딩_df_최종.head()
```
