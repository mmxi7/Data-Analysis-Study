3. 논리 연산(1) - 조건에 맞는 위치의 원소 추출

```python
# 철수의 1~4학년까지의 평균 점수
grade = np.array([82, 77, 91, 88])
grade
```

```python
# -> 추출하고자 하는 원본과 같은 사이즈의 배열을 생성
# -> 추출할 값은 True, 그렇지 않으면 False
bool_array = np.array([True, False, True, False])
bool_array
```

```python
# -> 조건에 맞는 항목만 1차 배열로 추출
result1 = grade[bool_array]
result1
```

4. 논리 연산(2) - 조건식에 맞는 원소 추출

```python
# 80점 이상은 원소만 추출
# -> 조건을 명시하기 위해 배열의 이름이 조건식에 명시된다
mypoint = np.array([82, 77, 91, 88])
ge80 = mypoint[mypoint >= 80]
ge80
```

```python
ge80le90 = mypoint[np.logical_and(mypoint >= 80, mypoint <= 90)]
ge80le90
```

```python
gt_or_lt = mypoint[np.logical_or(mypoint < 80, mypoint > 90)]
gt_or_lt
```

### #08 조건 분기

1. if~else 효과

```python
exam = np.array([60, 83, 75, 92, 86])

# 첫 번째 파라미터로 주어진 조건이 참일 경우 '합격',
# 그렇지 않을 경우 '불합격'
result = np.where(exam >= 80, '합격', '불합격')
result
```

2. 다중 조건 분기

```python
exam = np.array([60, 85, 51, 79, 94, 88])

# 학점을 부여하기 위한 점수의 구간을 설정하는 조건들을 리스트로 설정
conditions = [(exam >= 90),     # if
              (exam >= 80),     # elif
              (exam >= 70)      # elif
]

# 조건에 따라 부여될 학점
value = ['A', 'B', 'C']

# 조건과 조건을 판단하기 위한 값을 저장함
# default는 조건에 맞는 값이 아닌 경우 사용됨 (생략 가능)
result = np.select(conditions, value, default='F')
result
```

## 연습문제

□ 다음 조건을 모두 만족하는 파이썬 코드를 작성하시오.

NumPy의 ndarray를 활용하여

1. 정수 10, 20, 30, 40, 50으로 구성된 배열을 생성하고,
2. 배열의 길이(원소 개수)를 출력하며,
3. 배열의 첫 번째 원소와 마지막 두 개 원소를 각각 출력하고,
4. for문과 enumerate()를 이용해 인덱스와 값을 함께 출력하시오.

```python
arr = np.array([10, 20, 30, 40, 50])
size = len(arr)

print(f"배열의 길이: {size}")
print(f"첫 번째 원소: {arr[0]}")
print(f"마지막 두 개 원소: {arr[-2:]}")

for i, v in enumerate(arr):
    print("%d번째 원소 >> %d" % (i, v))
```
