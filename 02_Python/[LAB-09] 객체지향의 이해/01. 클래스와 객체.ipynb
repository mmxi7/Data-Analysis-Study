### #01 멤버변수만 포함하는 클래스

1. 클래스 정의하기

```python
# 변수들을 그룹화한 클래스 정의
class Dragoon:
    speed = None
    hp = None
    attack = None
    defense = None
```

2. 클래스로부터 객체 생성

```python
# 객체 생성하기
dg1 = Dragoon()

# 객체의 속성 설정하기
dg1.speed = 10
dg1.hp = 100
dg1.attack = 20
dg1.defense = 5

# 객체의 속성 일괄 출력하기
tpl = "speed: {0}, hp: {1}, attack: {2}, defense: {3}"
print(tpl.format(dg1.speed, dg1.hp, dg1.attack, dg1.defense))
```

3. 객체 안에 포함된 값 변경하기

```python
# 레벨업을 할 경우 모든 능력치가 증가한다
dg1.speed += 1
dg1.hp += 10
dg1.attack += 2
dg1.defense += 1

# 공격을 받을 경우 hp가 감소한다
dg1.hp -= 10

# 객체의 속성 일괄 출력
tpl = "speed: {0}, hp: {1}, attack: {2}, defense: {3}"
print(tpl.format(dg1.speed, dg1.hp, dg1.attack, dg1.defense))
```

4. 클래스의 재사용

```python
# 동일한 속성을 갖는 객체를 추가로 생성 가능
# 각 속성은 독립적인 값을 갖는다
dg2 = Dragoon()
dg2.speed = 12
dg2.hp = 120
dg2.attack = 22
dg2.defense = 7

tpl = "speed: {0}, hp: {1}, attack: {2}, defense: {3}"
print(tpl.format(dg1.speed, dg1.hp, dg1.attack, dg1.defense))
print(tpl.format(dg2.speed, dg2.hp, dg2.attack, dg2.defense))
```

### #02 메서드만 포함하는 클래스

1. 클래스 정의하기

```python
class Probe:

    # 이동메서드 -> 이동할 좌표를 전달받아 이동한다
    def move(self, x, y):
        print("좌표 ({0}, {1})로 이동합니다.".format(x, y))

    # 광물 채집 메서드 -> 채집된 광물을 반환한다
    def mineral(self):
        print("광물을 채집합니다.(+10)")
        return 10

    # 가스 채집 메서드 -> 채집된 가스를 반환한다
    def gas(self):
        print("가스를 채집합니다.(+5)")
        return 5
```

2. 객체를 통한 메서드 호출

```python
probe1 = Probe()
probe1.move(10, 10)

my_mieral = 0
my_mieral += probe1.mineral()
my_mieral += probe1.mineral()
my_mieral += probe1.mineral()

my_gas = 0
my_gas += probe1.gas()
my_gas += probe1.gas()

print("채취된 광물: {0}, 채취된 가스: {1}".format(my_mieral, my_gas))
```
