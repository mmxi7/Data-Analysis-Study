### 성적 처리 시스템 만들기

당신은 한 학급의 성적 데이터를 NumPy로 분석해야 하는 데이터 분석가입니다.   

학생 6명의 점수 데이터가 다음과 같이 주어집니다.   

score = [72.5, 88.0, 91.5, 67.0, 80.0, 59.5]   

다음 조건에 따라 코드를 작성하세요.   
- 위 점수 리스트를 정수형 ndarray로 변환하세요.
- 점수가 80점 이상인 학생의 점수만 추출하여 출력하세요.
- 모든 학생에게 다음 기준으로 학점을 부여하고 결과를 출력하세요.
- 90점 이상 → ‘A’ / 80점 이상 → ‘B’ / 70점 이상 → ‘C’ / 그 외 → ‘F’

```python
import numpy as np

score = np.array([72.5, 88.0, 91.5, 67.0, 80.0, 59.5], "int")
ge80 = score[score >= 80]

conditions = [(score >= 90),     # if
              (score >= 80),     # elif
              (score >= 70)      # elif
]

value = ['A', 'B', 'C']
result = np.select(conditions, value, default='F')

print("원본 점수:", score)
print("80점 이상:", ge80)
print("학점 결과:", result)
```

### 2. AI 교육생 점수 데이터 점검하기

당신은 AI 교육기관에서 수강생들의 모의시험 데이터를 검수하는 일을 맡고 있습니다.   
시험 데이터는 3회차 시험 × 4명 학생의 점수로 구성되어야 합니다.   
다음 조건에 맞게 NumPy로 코드를 작성하세요.   

평균이 75, 표준편차가 5인 정규분포를 따르는 랜덤 점수 데이터를 3행 4열 배열로 생성하십시오.   
단, 결과가 재현되도록 랜덤 시드를 100으로 고정해야 함.   
생성된 점수 중 80점 이상인 항목만 추출하시오.   
점수가 저장된 배열의 차원(ndim), 형태(shape), 데이터 타입(dtype) 을 각각 출력하시오.

```python
import numpy as np

np.random.seed(100)
score = np.random.normal(75, 5, (3, 4))

ge80 = score[score >= 80]

print(score)
print("80점 이상 점수:", ge80)
print("차원:", score.ndim)
print("형태:", score.shape)
print("데이터 타입:", score.dtype)
```
